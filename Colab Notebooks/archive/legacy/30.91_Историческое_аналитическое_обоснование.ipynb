# Аналитическое обоснование негативных результатов (без данных)

Чисто модельный расчёт: почему при толстых тканях ρ₂ неидентифицируем, облако решений разваливается, а Δρ₁ ловит ошибку модели. Всё выводится из прямой формулы и её производных — **без экспериментальных данных**. Это мост к МКЭ (после — расчёт по конечно-элементным моделям в MATLAB).

**Что доказываем:**
1. вся обусловленность зависит только от безразмерной глубины η=h/a (§1);
2. чувствительность к ρ₂ падает как ~1/η³ при толстых тканях (§2);
3. неопределённость ρ₂ (Крамер–Рао) взрывается с h — у Георга std(ρ₂)≫ρ₂ (§3);
4. усиление ошибки h→ρ₂ и развал облака (угол кривих → 0) (§4);
5. по одной сборке параметры не разделить; систолический split обусловлен L-формой, но ошибка модели по глубине протекает в Δρ₁ (§5);
6. критический размер L∝h: диапазон 50–140 для Георга ниже порога (§6).

## §0. Назначение анализа и словарь терминов

### Решаемая задача
Требуется показать аналитически, без обращения к данным, что негативные результаты работы (неопределимость удельного сопротивления лёгкого ρ₂ при толстом слое мягких тканей, широкий разброс оценки и появление ложного пульсового приращения мягких тканей Δρ₁) — не дефекты данных, а следствия свойств самой двуслойной модели. Этот раздел служит эталоном для сравнения с последующим конечноэлементным расчётом по реальной геометрии.

### Метод и обоснование выбора
Используются свойства прямой задачи двуслойной модели: масштабная инвариантность (форма зависимости определяется только безразмерной глубиной), асимптотическое поведение чувствительности и граница Крамера–Рао для неопределённости оценки. Аналитический подход выбран потому, что он отделяет ограничения модели от особенностей конкретных измерений.

### Словарь терминов
Основные термины (двуслойная модель, электродная сборка, эластичность, показатель качества, матрица Фишера) определены в §0 ноутбуков [09](33.01_Статическая_оценка_параметров_боковых_сборок.ipynb) и [05](30.04_Вычислительное_ядро_двуслойной_модели.ipynb). Дополнительно:
- **Безразмерная глубина** η = h/a — отношение толщины слоя мягких тканей к токовой полубазе сборки; определяет форму зависимости импеданса от параметров.
- **Масштабная инвариантность** — свойство модели, при котором приведённый импеданс и эластичности зависят только от безразмерной глубины, формы сборки и контраста, но не от абсолютных размеров по отдельности.
- **Коллапс чувствительности** — быстрое (примерно как обратный куб безразмерной глубины) убывание эластичности по ρ₂ с ростом глубины; означает потерю информации о лёгком при толстом слое мягких тканей.
- **Граница Крамера–Рао** — нижняя граница дисперсии несмещённой оценки, вычисляемая по матрице Фишера ([вторичный источник](https://ru.wikipedia.org/wiki/Неравенство_Крамера_—_Рао)).
- **Критический размер сборки** — наименьший размер, при котором эластичность по ρ₂ достигает порога; растёт линейно с толщиной слоя мягких тканей.

### Формулы
Безразмерная глубина, формула (12.1):

$$\eta=\frac{h}{a} \tag{12.1}$$

где η — безразмерная глубина; h — толщина слоя мягких тканей, м; a — токовая полубаза сборки, м.

Масштабная инвариантность импеданса, формула (12.2):

$$Z=\frac{\rho_1}{a}\,\Phi(\beta,\eta,k) \tag{12.2}$$

где Φ — безразмерная функция; β = b/a — форма сборки; k = (ρ₂ − ρ₁)/(ρ₂ + ρ₁) — контраст. Форма зависимости и эластичности определяются только (β, η, k).

Асимптотический коллапс чувствительности к лёгкому, формула (12.3):

$$S_{\rho_2}\ \sim\ \frac{1}{\eta^{3}}\quad (\eta\gg1) \tag{12.3}$$

где S_ρ₂ — эластичность импеданса по ρ₂ (§0 [05](30.04_Вычислительное_ядро_двуслойной_модели.ipynb)). При большой глубине чувствительность к лёгкому убывает примерно как обратный куб глубины.

**Входные данные.** Рабочая точка анализа: удельные сопротивления ρ₁ = 6, ρ₂ = 18 Ом·м; данные не используются.

**Допущения.** Импеданс, производные и эластичности двуслойной модели заданы аналитически (§0 [05](30.04_Вычислительное_ядро_двуслойной_модели.ipynb)); анализ самодостаточен и не зависит от измерений.

```python
# @title Модель Z, производные, эластичности (самодостаточно)
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
NT=400
def _terms(r1,r2,h,a,b,N=NT):
    k=(r2-r1)/(r1+r2); d1,d2=a-b,a+b; i=np.arange(1,N+1)
    G1=1/np.sqrt(d1**2+(2*i*h)**2); G2=1/np.sqrt(d2**2+(2*i*h)**2)
    return k,d1,d2,i,G1,G2,G1-G2
def Z(r1,r2,h,a,b,N=NT):
    k,d1,d2,i,G1,G2,dG=_terms(r1,r2,h,a,b,N)
    return r1/np.pi*(1/d1-1/d2)+2*r1/np.pi*np.sum(k**i*dG)
def dZ_dr2(r1,r2,h,a,b,N=NT):
    k,d1,d2,i,G1,G2,dG=_terms(r1,r2,h,a,b,N); T=np.sum(i*k**(i-1)*dG)
    return 4*r1**2/(np.pi*(r1+r2)**2)*T
def dZ_dr1(r1,r2,h,a,b,N=NT):
    k,d1,d2,i,G1,G2,dG=_terms(r1,r2,h,a,b,N); T=np.sum(i*k**(i-1)*dG)
    return Z(r1,r2,h,a,b,N)/r1-4*r1*r2/(np.pi*(r1+r2)**2)*T
def dZ_dh(r1,r2,h,a,b,N=NT):
    k,d1,d2,i,G1,G2,dG=_terms(r1,r2,h,a,b,N); U=np.sum(i**2*k**i*(G1**3-G2**3))
    return -8*r1*h/np.pi*U
def S_r2(r1,r2,h,a,b): return r2/Z(r1,r2,h,a,b)*dZ_dr2(r1,r2,h,a,b)
def S_h(r1,r2,h,a,b):  return h /Z(r1,r2,h,a,b)*dZ_dh(r1,r2,h,a,b)
def ab(L_mm,beta=0.5): a=L_mm/2/1000.0; return a,beta*a

# базовая рабочая точка (физиологичная, не из данных)
R1,R2=6.0,18.0
print("Модель и производные готовы. Рабочая точка ρ1=%.0f, ρ2=%.0f Ом·м." % (R1,R2))
```

**Анализ результатов.** Модель, производные и эластичности определены в рабочей точке.

**Результаты и умозаключения.** Эти функции — общий вход для всех разделов анализа.

## §1. Форма зависимости определяется безразмерной глубиной $\eta=h/a$

**Входные данные.** Приведённый импеданс и эластичность по ρ₂ при разных значениях безразмерной глубины (12.1) и двух масштабах абсолютного размера.

**Допущения.** Чтобы проверить масштабную инвариантность (12.2), приведённый импеданс и эластичность вычисляются при одинаковой безразмерной глубине, но разных абсолютных размерах: они должны совпасть.

```python
# @title Проверка масштабной инвариантности (Z·a/ρ1 и S_ρ2 зависят только от η,k,β)
print("η     | Z·a/ρ1 (масштаб ×1, ×3)        | S_ρ2 (×1, ×3)")
for eta in [0.3,0.6,1.0]:
    for a_mm,tag in [(50,"×1"),(150,"×3")]:
        a=a_mm/1000.0; b=0.5*a; h=eta*a
        za=Z(R1,R2,h,a,b)*a/R1; s=S_r2(R1,R2,h,a,b)
        if tag=="×1": za1,s1=za,s
    print("%.1f   | %.5f / %.5f (совпали: %s) | %.4f / %.4f" %
          (eta, za1, za, np.isclose(za1,za,rtol=1e-6), s1, s))
print("\nВывод: форма зависимости и чувствительность — функции ТОЛЬКО η=h/a (и β,k). Толщина входит лишь через η.")
```

**Анализ результатов.** При изменении абсолютного масштаба в три раза приведённый импеданс и эластичность по ρ₂ не меняются (совпадают при каждом значении η). Форма зависимости и чувствительность — функции только безразмерной глубины, формы и контраста.

**Результаты и умозаключения.** Толщина входит в задачу только через безразмерную глубину η = h/a; поэтому любой порог чувствительности задаёт условие на η, а не на абсолютную толщину.

## §2. Чувствительность к $\rho_2$ убывает примерно как $1/\eta^{3}$

**Входные данные.** Эластичность по ρ₂ как функция безразмерной глубины; положения Ника и Георгия на этой зависимости.

**Допущения.** Чтобы показать потерю чувствительности к лёгкому при толстом слое, строится зависимость эластичности по ρ₂ от глубины и проверяется асимптотика (12.3).

```python
# @title S_ρ2(η): коллапс ~1/η³ + положение Ника и Георгия
eta=np.logspace(-1.0,0.8,80)
a=0.035  # фиксируем a, меняем h=η·a (форма от a не зависит, §1)
S=np.array([S_r2(R1,R2,e*a,a,0.5*a) for e in eta])
plt.figure(figsize=(8,5))
plt.loglog(eta,S,lw=2,label="$S_{ρ2}(η)$")
ref=S[40]*(eta/eta[40])**-3; plt.loglog(eta,ref,"k--",lw=1,label="наклон −3 (∝1/η³)")
# диапазоны η=h/a для решёток 50..140 (a=L/2)
for h_mm,name,c in [(15,"Ник",'tab:blue'),(40,"Георгий",'tab:red')]:
    e=[h_mm/(L/2) for L in [50,140]]
    plt.axvspan(min(e),max(e),alpha=0.15,color=c,label="%s: η=%.2f..%.2f"%(name,min(e),max(e)))
plt.xlabel("η = h/a"); plt.ylabel("$S_{ρ2}$ (доля сигнала от лёгкого)")
plt.title("Чувствительность к ρ₂ исчезает с глубиной как ~1/η³"); plt.legend(fontsize=8); plt.grid(True,which="both"); plt.tight_layout(); plt.show()
print("Георгий (толстые ткани) сидит в зоне высокого η -> S_ρ2 в разы меньше, чем у Ника.")
```

**Анализ результатов.** Эластичность по ρ₂ убывает с глубиной примерно как обратный куб; Георгий (толстый слой) находится в зоне большой глубины, где эластичность в разы меньше, чем у Ника.

**Результаты и умозаключения.** При толстом слое мягких тканей чувствительность импеданса к ρ₂ мала независимо от объекта; это первопричина неопределимости ρ₂ у Георгия ([09](33.01_Статическая_оценка_параметров_боковых_сборок.ipynb) §3.1).

## §3. Неопределённость $\rho_2$ растёт с толщиной (граница Крамера–Рао)

**Входные данные.** Матрица Фишера набора размеров 50–140 мм (§0 [05](30.04_Вычислительное_ядро_двуслойной_модели.ipynb)) при разной толщине.

**Допущения.** Чтобы связать чувствительность с неопределённостью оценки, стандартное отклонение оценки ρ₂ вычисляется по границе Крамера–Рао как функция толщины.

```python
# @title std(ρ2)/ρ2 vs толщина h — для набора 50..140 мм
EPS=0.01  # относительный шум измерителя, 1%
Ls=np.arange(50,141,10)
def std_rho2(h):
    G=np.array([[dZ_dr1(R1,R2,h,*ab(L)),dZ_dr2(R1,R2,h,*ab(L))] for L in Ls])
    sig=np.array([EPS*Z(R1,R2,h,*ab(L)) for L in Ls])
    M=(G/sig[:,None]).T@(G/sig[:,None])
    return np.sqrt(np.linalg.inv(M)[1,1])
hs=np.linspace(0.005,0.060,60)
sd=np.array([std_rho2(h) for h in hs])
plt.figure(figsize=(8,5))
plt.semilogy(hs*1000, sd/R2*100, lw=2)
for h_mm,name in [(15,"Ник"),(40,"Георгий")]:
    v=std_rho2(h_mm/1000)/R2*100
    plt.scatter([h_mm],[v],zorder=5); plt.annotate("%s: ±%.0f%%"%(name,v),(h_mm,v),fontsize=9)
plt.axhline(100,color="red",ls="--",lw=1,label="неопределённость = 100% ρ2 (предел осмысленности)")
plt.xlabel("толщина h, мм"); plt.ylabel("std($\\hatρ_2$)/ρ2, % (Крамер–Рао, шум 1%)")
plt.title("Неопределённость ρ2 взрывается с толщиной тканей"); plt.legend(); plt.grid(True,which="both"); plt.tight_layout(); plt.show()
print("При h=15 ρ2 определяется; при h=40 std(ρ2) сравнима/больше самого ρ2 -> неидентифицируем (как и в данных: упёрся в границу).")
```

**Анализ результатов.** При толщине 15 мм ρ₂ определяется (относительная неопределённость мала); при толщине 40 мм стандартное отклонение оценки сравнимо с самим ρ₂ или превышает его, то есть ρ₂ неидентифицируем.

**Результаты и умозаключения.** Неопределённость ρ₂ растёт с толщиной вплоть до полной неопределимости; это количественно объясняет выход оценки на границу у Георгия ([09](33.01_Статическая_оценка_параметров_боковых_сборок.ipynb) §3.1).

## §4. Усиление ошибки толщины и расширение облака (угол между кривыми уменьшается)

**Входные данные.** Показатель качества сигнал/помеха и угол между кривыми равного импеданса по размерам сборок как функция толщины.

**Допущения.** Чтобы объяснить расширение облака пересечений, вычисляются показатель качества (§0 [05](30.04_Вычислительное_ядро_двуслойной_модели.ipynb)) и угол между кривыми; малый угол усиливает разброс пересечений.

```python
# @title FoM(h) и угловой спред кривих по сборкам
def spread_angle(h):
    ang=[np.degrees(np.arctan2(dZ_dr2(R1,R2,h,*ab(L)), dZ_dr1(R1,R2,h,*ab(L)))) for L in Ls]
    return max(ang)-min(ang)
def fom(h,L): s2=S_r2(R1,R2,h,*ab(L)); sh=S_h(R1,R2,h,*ab(L)); return s2/abs(sh)
fig,ax=plt.subplots(1,2,figsize=(14,5))
ax[0].plot(hs*1000,[fom(h,140) for h in hs],lw=2)
ax[0].axhline(1,color="red",ls="--",label="FoM=1 (10% h -> 10% ρ2)")
for h_mm,n in [(15,"Ник"),(40,"Георгий")]: ax[0].scatter([h_mm],[fom(h_mm/1000,140)],zorder=5,label="%s"%n)
ax[0].set_xlabel("h, мм"); ax[0].set_ylabel("FoM = S_ρ2/|S_h| (L=140)"); ax[0].set_title("Сигнал/помеха падает с h"); ax[0].legend(); ax[0].grid(True)
ax[1].plot(hs*1000,[spread_angle(h) for h in hs],lw=2,color="tab:green")
for h_mm,n in [(15,"Ник"),(40,"Георгий")]:
    v=spread_angle(h_mm/1000); ax[1].scatter([h_mm],[v],zorder=5); ax[1].annotate("%s: %.0f°"%(n,v),(h_mm,v),fontsize=9)
ax[1].set_xlabel("h, мм"); ax[1].set_ylabel("разброс направлений кривих, °")
ax[1].set_title("Угол кривих → 0 с толщиной → облако разваливается"); ax[1].grid(True)
plt.tight_layout(); plt.show()
print("Толстые ткани: FoM<1 (ошибка h усиливается) и угол кривих мал (1/sinθ велик) -> облако в разы шире.")
```

**Анализ результатов.** При толстом слое показатель качества меньше единицы (ошибка толщины усиливается), а угол между кривыми мал, поэтому разброс облака велик.

**Результаты и умозаключения.** Расширение облака у Георгия — следствие обусловленности (малого угла и низкого показателя качества), а не ошибки данных ([09](33.01_Статическая_оценка_параметров_боковых_сборок.ipynb) §4–5).

## §5. По одной сборке разделение невозможно; ошибка глубины источника даёт ложное $\Delta\rho_1$

**Входные данные.** Модельный систолический размах при разной глубине пульсового источника (источник смещается вглубь от однородного полупространства).

**Допущения.** Чтобы показать происхождение ложного Δρ₁, в модель вносится ошибка глубины источника, и по двуслойной модели восстанавливаются Δρ₁ и Δρ₂. При верной модели (нулевом смещении) Δρ₁ должно быть строго нулём.

```python
# @title Демонстрация: смещение источника по глубине -> ложное Δρ1
h0=0.025; Dr2_true=0.08    # истинная пульсация лёгкого
Ls=np.arange(50,141,10)
J=np.array([[dZ_dr1(R1,R2,h0,*ab(L)),dZ_dr2(R1,R2,h0,*ab(L))] for L in Ls])  # базис на номинальном h0
print("Δd, мм | восстановл. Δρ1 (ложное) | Δρ2 | трактовка")
print("-"*64)
for dd in [0,5,10,20]:
    # "истина": источник лёгкого эффективно глубже на Δd (другая L-форма)
    dZ_true=np.array([dZ_dr2(R1,R2,h0+dd/1000,*ab(L))*Dr2_true for L in Ls])
    sol,*_=np.linalg.lstsq(J,dZ_true,rcond=None)
    print("  %2d   |      %+.4f Ом·м       | %.4f | %s" %
          (dd, sol[0], sol[1], "Δρ1=0 (нет ошибки)" if dd==0 else "ошибка глубины -> ложное Δρ1≠0"))
print("\nВывод: при ВЕРНОЙ модели (Δd=0) Δρ1=0 точно. Любое несоответствие глубины источника")
print("плоской двуслойной модели проецируется в Δρ1 -> это систематический предел (нужен МКЭ).")
```

**Анализ результатов.** При нулевом смещении Δρ₁ = 0 точно. Любое смещение источника вглубь даёт ненулевое (ложное) Δρ₁ и уменьшает восстановленное Δρ₂.

**Результаты и умозаключения.** Ложное Δρ₁ возникает, когда пульсовой источник (сердце, крупные сосуды) лежит глубже однородного полупространства модели; это модельная систематика, а не пульсация мягких тканей ([11](33.03_Пульсовые_ансамбли_боковых_сборок.ipynb) §3.3). Устранение требует глубинно-разрешённой модели (конечноэлементный расчёт по томографии).

## §6. Критический размер сборки растёт с толщиной: диапазон 50–140 мм для Георгия недостаточен

**Входные данные.** Критический размер сборки (по порогу эластичности к ρ₂) как функция толщины.

**Допущения.** Чтобы оценить достаточность имеющихся размеров, вычисляется наименьший размер сборки, при котором эластичность по ρ₂ достигает порога; ожидается линейный рост с толщиной.

```python
# @title L для порога S_ρ2 vs толщина
S_STAR=0.25
def L_for_threshold(h):
    Lg=np.linspace(40,400,500)
    s=np.array([S_r2(R1,R2,h,*ab(L)) for L in Lg])
    idx=np.where(s>=S_STAR)[0]
    return Lg[idx[0]] if len(idx) else np.nan
hs2=np.linspace(0.005,0.060,40)
Lneed=np.array([L_for_threshold(h) for h in hs2])
plt.figure(figsize=(8,5))
plt.plot(hs2*1000,Lneed,lw=2,label="L для $S_{ρ2}≥%.2f$"%S_STAR)
plt.axhspan(50,140,alpha=0.15,color="green",label="доступный диапазон 50–140 мм")
for h_mm,n in [(15,"Ник"),(40,"Георгий")]:
    v=L_for_threshold(h_mm/1000); plt.scatter([h_mm],[v],zorder=5); plt.annotate("%s: нужно %.0f мм"%(n,v),(h_mm,v),fontsize=9)
plt.xlabel("толщина h, мм"); plt.ylabel("необходимый размер L, мм")
plt.title("Критический размер растёт линейно с h (L∝h)"); plt.legend(); plt.grid(True); plt.tight_layout(); plt.show()
print("Ник: порог достигается внутри 50–140. Георгий: нужно L много больше 140 -> текущими решётками ρ2 не взять.")
```

**Анализ результатов.** У Ника порог достигается внутри диапазона 50–140 мм; у Георгия требуется размер существенно больше 140 мм.

**Результаты и умозаключения.** Текущий диапазон размеров достаточен для тонкого слоя мягких тканей и недостаточен для толстого; это обосновывает расширение набора размеров вверх ([05](30.04_Вычислительное_ядро_двуслойной_модели.ipynb) §8) и служит эталоном для конечноэлементного расчёта (§7).

## §7. Выводы и мост к МКЭ

**Все негативные результаты — следствия двух фактов:**
1. $S_{\rho_2}\sim1/\eta^3$ — чувствительность к лёгкому исчезает кубически с относительной глубиной. Отсюда: взрыв std(ρ₂) с h (§3), падение FoM и развал облака (§4), критический $L\propto h$ за пределами доступного для толстых тканей (§6).
2. Двухпараметрическая плоская модель **рангово бедна** для глубокого/структурированного источника: одна сборка не разделяет (§5), а несоответствие глубины проецируется в Δρ₁ — систематический предел, не убираемый статистикой.

**Что снимает МКЭ по геометрии КТ (следующий шаг, MATLAB):**
- **глубинно-разрешённый источник** — пульсация на правильной глубине (сердце/сосуды/лёгкое) убирает протекание в Δρ₁ (§5);
- **реальная геометрия** (рёбра, кривизна) — даёт верхнюю границу $L_{\max}$ и позволяет крупные базы, возвращающие $S_{\rho_2}$ при толстых тканях (§2,§6);
- **проверка плоской формулы**: сравнение $Z_{\text{МКЭ}}(L)$ с аналитической $Z(L)$ количественно очерчивает зону применимости.

Эти аналитические зависимости (S_ρ2(η), std(ρ2)(h), FoM, L(h)) — эталон, с которым будем сверять МКЭ.
