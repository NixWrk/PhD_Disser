# 21.10. Сопоставление изменений объёма камер сердца на исторической контрастной КТ

## Резюме

Для одной исторической серии КТ сопоставлены два способа выделения полостей сердца на 20 фазах реконструкции. Первый способ использует контуры, заданные по КТ исходной фазы, их перенос по деформации изображения и пороговое выделение ярких вокселей. Второй способ — TotalSegmentator v2. Расчёт охватывает оба желудочка, оба предсердия, а также отдельные кандидатные маски наружной области сердца и ткани стенки.

**Основной результат пересмотра:** прежнее сравнение смешивало геометрическую область камеры и её яркое подмножество, а перенос контуров давал значительные ошибки разделения правых камер. После прямой перерисовки двух фаз и раздельного учёта этих определений расхождение изменений ПЖ уменьшилось с 62,26 до 4,98 мл, ПП — с 65,96 до 7,22 мл. Для ЛЖ прежняя близость оказалась неустойчивой: расхождение выросло с 4,24 до 16,44 мл. Эти числа характеризуют различие способов, а не установленную погрешность TotalSegmentator.

**Можно ли доверять числам TotalSegmentator:** вычислительный подсчёт объёмов проверен; точность анатомических объёмов и их изменений пока не установлена. Алгоритм остаётся источником исследовательских оценок. Наш прежний кандидат недостаточно надёжен для его валидации, а исправленные контуры двух фаз выполнены без ослепления и также не приняты как референс.

Расчёт всех 20 фаз завершён. Подробно просмотрены 96 аксиальных срезов двух фаз, исправлены кандидатные области камер и подготовлены редактируемые проекты. Разделы 3–5 сохраняют исходный расчёт для воспроизводимости; **актуальная анатомическая ревизия и её объяснение представлены в разделе 6**, итоговые ограничения — в разделе 7. Полная проверка остальных фаз, миокарда и наружной границы остаётся незавершённой.

## 1. Цель и место исследования

Задача состоит в проверке того, насколько результат TotalSegmentator воспроизводит изменение объёмов камер на протяжении реконструированного сердечного цикла. Для такой проверки нужен второй способ разметки с известными анатомическими границами и достаточной независимостью от проверяемого алгоритма. Совпадение объёмов само по себе не доказывает совпадение границ.

Предыдущий отчёт 21.09 описывает исправление частичных областей желудочков. Здесь рассматривается следующий этап: расширенные области четырёх камер и их динамика. Данные относятся к одной выбранной серии исторической КТ; результаты нельзя переносить на остальные исторические серии или на современные КТ добровольцев без отдельного сравнения.

## 2. Исходные данные и определения

Использованы 20 объёмов одной контрастной КТ с фазами от 0 до 95% через 5%. Размер массива каждой фазы — 512 × 512 × 103, шаг отсчётов — приблизительно 0,473 × 0,473 × 2,5 мм. Указанная в исходных метаданных толщина среза составляет 5 мм. Шаг 2,5 мм не означает независимое пространственное разрешение 2,5 мм: соседние реконструированные срезы перекрываются. Пересчёт в изотропные воксели не восстановит отсутствующие детали клапанов и тонких стенок.

Обозначения: ЛЖ — левый желудочек, ПЖ — правый желудочек, ЛП — левое предсердие, ПП — правое предсердие. Под объёмом в вычислительных таблицах понимается объём конкретной выделенной области. Для исходного порогового кандидата это яркое подмножество поисковой области; для TotalSegmentator — вся метка камеры; для исправленных прямых контуров — непороговая геометрическая область. Анатомическая эквивалентность этих определений не предполагается. Сумма четырёх масок не включает отдельную метку миокарда и не равна объёму всего сердца; отнесение внутриполостных мышечных структур требует отдельного правила. Наружная область сердца и ткань его стенки рассматриваются отдельно.

Процент фазы задаёт положение в последовательности реконструкции. Фаза 0% выбрана как начало отсчёта изменений, а не как доказанная конечная диастола. Конечная диастола и конечная систола независимо не установлены; ударный объём и фракция выброса здесь не рассчитываются.

|---|---|---|

| Контуры исходной фазы | Заданы агентом по разрешённому просмотру обезличенной КТ | Кандидатная анатомическая разметка; экспертно не принята |
| Другие фазы в исходном варианте | Вычислены по деформации КТ и локальному порогу яркости | Результат алгоритма, без независимой ручной обводки каждой фазы |
| Исправленные прямые контуры фаз 0 и 40% | Заданы по КТ без межфазного переноса | Разметка агентом после просмотра TotalSegmentator; независимость отсутствует |

## 3. Исходные способы сегментации и введённые допущения

### 3.1. Контуры исходной фазы

Области желудочков были расширены относительно частичных масок отчёта 21.09; дополнительно заданы области предсердий. Опорные контуры интерполированы между срезами. Разделение желудочков и предсердий включает приближённые плоскости клапанных границ. В области правых камер разделяющее правило действует в пересечении их поисковых областей, чтобы не отсечь весь передний выходной тракт одним глобальным сечением.

Границы клапанов, переходы к магистральным сосудам, устья вен и полнота включения ушек предсердий остаются неопределёнными. В частности, верхняя граница кандидатного ПЖ в исходной фазе достигает среза 35, тогда как автоматическая маска продолжается выше. Это различие требует просмотра клапанной области в нескольких плоскостях. Нельзя заранее считать дополнительную область TotalSegmentator сосудом, ошибочно включённым в желудочек.

### 3.2. Перенос на остальные фазы

Изображение каждой целевой фазы совмещено с исходной КТ посредством B-сплайнового поля деформации. При оценке преобразования использованы только изображения КТ; маски TotalSegmentator не участвовали. Вычисление выполнено в ограниченной области сердца на сетке приблизительно 2 × 2 × 2,5 мм, с расстоянием между управляющими узлами 20 мм, тремя уровнями разрешения и лимитом 40 итераций на уровень. Критерий совмещения — взаимная информация.

Преобразование переносит исходные поисковые области на целевую КТ. В каждой области выделяются воксели с интенсивностью не ниже 190 HU; сохраняется крупнейшая компонента с шестисвязным соседством. Дополнительно рассчитаны пороги 170 и 210 HU. Это выбранные вычислительные сценарии, а не литературные границы нормальной плотности крови и не доверительный интервал ошибки. Одинаковый порог не гарантирует одинаковой анатомической полноты при изменении контраста и частичного объёма.

Отдельно сохранён перенос готовых бинарных масок без повторного выделения по интенсивности. Сравнение двух вариантов позволяет оценить, насколько объём определяется деформацией исходной области и насколько — интенсивностью целевой КТ. Ни один вариант не считается эталоном только по внутренней согласованности.

### 3.3. Наружная область и кандидат ткани стенки

Наружный контур задан по исходной КТ и перенесён на остальные фазы. Его границы у основания и сосудов пока не приняты. Из области внутри контура исключены четыре кандидатные полости. Полученный остаток дополнительно ограничен интенсивностью от 0 до 190 HU. В отчёте он назван кандидатом ткани стенки.

Такой остаток может содержать невыделенную кровь, соединительную ткань, участки сосудов и ошибки контура. Поэтому он не эквивалентен морфологически проверенному миокарду. На последующих фазах эта исходная маска переносится как бинарная область; порог ткани заново не применяется. Кривую этого кандидата можно сопоставить с маской миокарда TotalSegmentator только для диагностики различий определений.

### 3.4. Независимость и контроль вычислений

Исходные контуры и правила переноса были зафиксированы до просмотра численного сравнения и автоматических контуров. После этого выявлялись пространственные причины расхождений. Любая следующая коррекция, сделанная с учётом уже увиденных масок TotalSegmentator, должна быть отмечена как разметка без ослепления. Она не становится независимым референсом автоматически.

Перед сравнением проверяются контрольные суммы исходных КТ и масок, соответствие фаз, размеры массивов, аффинные преобразования и единицы миллиметров. Маски должны быть бинарными, непустыми и не пересекаться между камерами одного способа. Эти проверки защищают от подмены серии, ошибки масштаба и двойного подсчёта. Они не устанавливают правильность анатомических границ.

## 4. Расчёт показателей

Объём выделенной области определяется суммой физических объёмов её вокселей:

$$V_c(t)=N_c(t)\,|\det A|/1000,$$

где $N_c(t)$ — число вокселей камеры $c$ на фазе $t$, $A$ — пространственная часть аффинного преобразования в миллиметрах. Деление на 1000 переводит мм³ в мл.

Изменение относительно исходной фазы и расхождение изменений вычисляются как

$$\Delta V_c(t)=V_c(t)-V_c(0),\qquad D_c(t)=\Delta V_c^{TS}(t)-\Delta V_c^{C}(t),$$

где индексы $TS$ и $C$ обозначают TotalSegmentator и кандидатную разметку соответственно. Положительное $D$ означает, что TotalSegmentator даёт большее увеличение или меньшее уменьшение объёма от собственного исходного уровня. Отрицательное значение означает обратное. Постоянное смещение объёма между способами вычитается при расчёте $D$; изменяющееся по фазам смещение сохраняется.

Рисунок 4. Перенос готовой бинарной маски и повторное выделение по интенсивности в перенесённой поисковой области. Оба варианта используют одно поле деформации КТ.

На фазе 50% объём ЛЖ при переносе готовой маски равен 73.27 мл, а после порогового выделения в целевой КТ — 54.32 мл. Различие составляет 18.94 мл. Оно получено на одном изображении и поэтому характеризует зависимость оценки от алгоритма обработки, а не дополнительное физиологическое сокращение.

Пороговое выделение удаляет часть низкоинтенсивных участков, однако не исправляет ошибочное отнесение яркого участка к другой камере. Приближённая клапанная граница и ошибка поля деформации сохраняются. Близость одного из вариантов к TotalSegmentator не является достаточным основанием для его выбора.

Эти наблюдения согласуются с недостатками исходной границы и её переноса. Поэтому большое расхождение динамики ПЖ нельзя целиком приписывать TotalSegmentator. Для решения нужны разметка клапанного кольца и прослеживание выходного тракта в согласованных продольных и поперечных плоскостях.

Сохранённые маски позволяют просматривать внешнюю геометрию и отделять полости от окружающего остатка. Однако верхнее отсечение наружного контура и низкая тканевая специфичность порогового остатка не позволяют считать задачу сегментации миокарда завершённой. ## 6. Пересмотр результатов после прямого анатомического просмотра

### 6.1. Объём полости и объём вокселей с заданной интенсивностью

При повторной проверке обнаружено принципиальное ограничение исходного сравнения. Автоматическая маска камеры и область, выделенная порогом 190 HU внутри приближённого контура, задают разные измеряемые величины. Первая представляет присвоенную алгоритмом анатомическую область. Вторая содержит только достаточно яркие воксели внутри поисковой области. Их различие зависит от фазы, поэтому вычитание исходного объёма не устраняет его.

Для проверки этого эффекта на всех 20 фазах сохранены неизменными маски TotalSegmentator и отдельно подсчитаны все их воксели и подмножества с интенсивностью не ниже 170, 190 и 210 HU. Тем самым влияние порога исследовано без перемещения границ маски:

$$V_{190}(t)=v\sum_{x\in M(t)}\mathbf{1}\{I(x,t)\geq190\},\qquad V_M(t)=v|M(t)|,$$

где $M(t)$ — автоматическая маска камеры, $I(x,t)$ — интенсивность КТ в HU, а $v$ — физический объём вокселя в мл. Величина

$$B_{190}(t)=[V_{190}(t)-V_{190}(0)]-[V_M(t)-V_M(0)]$$

характеризует изменение результата, вызванное только введением порога. Это диагностический эффект определения объёма; его нельзя интерпретировать как установленную ошибку сегментации крови.

| ЛЖ | 145.83 → 61.79 | 136.43 → 38.66 | -13.72 |
| ПЖ | 196.48 → 107.39 | 179.39 → 73.52 | -16.78 |

| ПП | 62.05 → 128.89 | 54.80 → 114.73 | -6.91 |



Рисунок 10. Межфазные изменения внутри одних и тех же масок TotalSegmentator: все воксели и только воксели с интенсивностью не ниже 190 HU. Расхождение кривых вызвано пороговой обработкой при неизменных анатомических метках.

Для ЛЖ на фазе 40% порог добавляет 13,72 мл к оценке уменьшения объёма; для ПЖ — 16,78 мл. На всём цикле максимальный модуль этого эффекта достигает 18,67 мл для ЛЖ и 16,78 мл для ПЖ. Следовательно, даже визуально устойчивый порог может существенно искажать оцениваемую амплитуду изменения. Эти числа получены внутри масок TotalSegmentator и не являются поправками, которые можно автоматически вычесть из прежней кривой кандидата.

В маске ЛЖ доля вокселей ниже 170 HU возрастает с 3,0% на исходной фазе до 31,8% на фазе 40%. Низкая интенсивность может быть связана с трабекулами и сосочковыми мышцами, частичным объёмом, неоднородностью контрастирования или ошибками контура. По одному порогу эти причины не разделяются. Поэтому ни все воксели автоматической маски, ни только яркое подмножество нельзя без дополнительной анатомической проверки объявить истинным объёмом крови.

Для оценки функции сердца требуется единое правило включения сосочковых мышц и трабекул во всех фазах и обоих способах разметки. Для электрического моделирования дополнительно требуется различать кровь и ткань миокарда: геометрическая полость с включёнными внутриполостными мышечными структурами не является однородным кровяным компартментом. Совпадение клинической условности обводки и физического определения компартмента пока не установлено.

### 6.2. Исправление прямых контуров двух фаз

Последовательно просмотрены аксиальные срезы 2–49 фаз 0 и 40%: по 48 срезов каждой фазы, сначала с автоматическими контурами, затем с исправленными кандидатными контурами. Это подробный просмотр двух объёмов, а не анатомическая проверка всех 20 фаз. Область выше среза 49 и остальные 18 фаз не размечены вручную заново.

Просмотр подтвердил два недостатка прежнего кандидата. На исходной фазе верхнее ограничение ПЖ срезом 35 исключало часть его переднего выходного отдела. На фазе 40% перенос исходной области относил значительную часть правого предсердия к ПЖ. По пространственному пересечению с автоматической маской ПП объём такого несогласованного участка составлял 37,98 мл. Это локализация расхождения методов, а не автоматически установленный объём ошибочно размеченной ткани.

Для исходной фазы заново заданы контуры ПЖ и ПП; прежние непороговые области ЛЖ и ЛП сохранены. Для фазы 40% заново заданы контуры всех четырёх камер непосредственно по её КТ. Межфазное поле деформации при построении этих новых областей не использовалось. Непороговый объём внутри контура и результаты трёх HU-порогов сохранены раздельно. Неоднозначные пересечения контуров исключены из всех камер и сохранены отдельной маской: 0,138 мл в исходной фазе и 0,366 мл на фазе 40%. Это только объём пересечений, а не полная неопределённость разметки.

Новые контуры создавались после просмотра TotalSegmentator. Поэтому они представляют исправление без ослепления, выполненное агентом, и не являются независимым экспертным референсом. Уровни клапанных отсечений, устья сосудов, ушки предсердий и внутриполостные мышечные структуры остаются источниками неопределённости. Полигональное приближение также ограничивает точность поверхности.

### 6.4. Миокард и наружная граница

Повторный просмотр автоматических контуров показал, что метка миокарда на выбранных срезах охватывает преимущественно стенку ЛЖ; непрерывного выделения всей свободной стенки ПЖ и стенок обоих предсердий на этих изображениях не видно. В нашей разметке прежний остаток внутри наружного контура также не стал анатомическим миокардом: он содержит неопределённость полостей и основания сердца. Наружная область и миокард сохранены как отдельные черновики; их валидация не завершена. Новые прямые контуры двух фаз исправляют камеры и не подтверждают автоматически эти две дополнительные структуры.

### 6.5. Применимость опубликованных характеристик TotalSegmentator

В [описании модели камер высокого разрешения разработчиком](https://raw.githubusercontent.com/wasserth/TotalSegmentator/master/resources/heartchambers_highres_details.md) указаны 1502 обучающих и 57 валидационных изображения, рабочее разрешение приблизительно 0,73 × 0,72 × 1,0 мм и средний Dice 0,950 по классам в валидационной выборке. Эти сведения относятся к конкретной модели и её собственной выборке.

Коэффициент Dice характеризует пространственное перекрытие масок. Значение 0,950 не означает погрешность объёма 5% и не устанавливает точность межфазных разностей. Здесь исходная толщина среза 5 мм, а проверяется динамика одной исторической контрастной серии. Интерполяция к рабочей сетке модели не устраняет это различие исходных данных. Поэтому опубликованную характеристику нельзя использовать вместо локальной проверки точности объёмов.

## 7. Достоверность результатов и условия применения

Повторная проверка изменила интерпретацию исходного расчёта. Большие расхождения правых камер во многом связаны с дефектами нашей поисковой области и её переноса. Малое прежнее расхождение ЛЖ также неустойчиво: на него влияет определение измеряемого объёма. Следовательно, исходные 20-фазные кривые полезны для диагностики метода, но недостаточны для количественной валидации TotalSegmentator.

| Проверяемое свойство | Что установлено | Что ещё не установлено |
|---|---|---|
| Подсчёт объёма по сохранённой маске | Совпадение КТ, физических сеток, единиц и контрольных сумм; воспроизводимые межфазные разности | Анатомическая правильность самой маски |
| Основные камеры на двух просмотренных фазах | Контуры TotalSegmentator распознают соответствующие крупные области; новая обводка устраняет грубые дефекты прежнего кандидата | Точность клапанных границ и полных поверхностей, все остальные фазы |
| Изменение объёмов | Рассчитано на 20 фазах; выявлены зависимость от определения и дефекты переноса | Погрешность относительно независимого референса |
| Миокард и наружная область | Сохранены автоматические маски и отдельные черновые области | Полнота охвата всей сердечной стенки и достоверность её объёма |
| Ударный объём и фракция выброса | В этом сравнении не рассчитывались | Независимый выбор конечных фаз и погрешность функциональных показателей |

**Практическое решение:** сохранить автоматические числа как исследовательские результаты с указанным происхождением, но не использовать прежнее согласие с нашим кандидатом как свидетельство их точности. Не вводить по этим расхождениям постоянные поправки к камерам и не заменять автоматические маски в основном расчёте исправленными кандидатами. Для электрического моделирования отдельно согласовать геометрическое определение крови и миокарда.

После независимой разметки можно будет оценить абсолютные расхождения, различия межфазных изменений и поверхностей, а также их воспроизводимость между разметчиками. Допустимую погрешность следует задать исходя из требуемой точности последующей оценки функции сердца или электрической модели. Сейчас такой численный допуск для этой проверки не утверждён; выбирать его после просмотра результата означало бы подгонять критерий.

## 8. Дальнейшая анатомическая проверка

1. На исходной фазе определить митральное и трёхстворчатое кольца, аортальный и лёгочный клапаны в нескольких согласованных плоскостях. Сохранить координаты, правило отсечения и неоднозначные участки. Отдельно проверить устья вен и ушки предсердий.
2. Исправить исходные границы ПЖ, предсердий и наружной области по этой анатомической разметке. Зафиксировать, что после настоящего сравнения разметчик уже видел результат TotalSegmentator; прежний вариант сохранить для истории.
