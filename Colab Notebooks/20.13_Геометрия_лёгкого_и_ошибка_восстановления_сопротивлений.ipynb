## 2. Пространственные сокращения

Проксимальная часть лёгочной маски выделяется плоскостью через середину диапазона глубин вдоль $d$. Это геометрическое правило, не анатомическое деление на правое и левое лёгкое. Центр и отношения полуосей эллипсоида определяются по объёмно-взвешенным моментам тетраэдров; непрерывный объём приводится к объёму выделенной части. Пересечение с телом и объединение с сохранённой дистальной маской могут изменить итоговый объём. Привязка по глубине переносит эллипсоид вдоль $d$ до совпадения центрального входа и не использует импеданс.

Плоская пространственная модель заменяет проксимальную область полупространством за центральным входом; наружное тело сохраняется. Продолженная маска заполняет лёгким весь участок луча после первого входа внутри тела. Она удаляет все дистальные выходы; это не изолированное устранение купола диафрагмы.

Центральный вход в дискретную маску также отличается от пересечения исходной поверхности: эти величины нельзя отождествлять без проверки сетки.

## 3. Слоистые приближения

Сохраняются кожа и все упорядоченные входы и выходы лёгкого каждого луча. Новые варианты используют табличный профиль C01, без переноса прежнего полинома. Луч без лёгкого получает однородное ядро. В варианте среднего $h$ такой участок делает среднее бесконечным и устраняет лёгочную поправку данной пары. Это свойство выбранной эвристики, а не отсутствие чувствительности к соседнему лёгкому.

В варианте среднего первого интервала глубина и толщина усредняются только по лучам с лёгким. Слоистое ядро взвешивается по их доле, остальная доля получает однородное ядро. Такое правило задано до электрической подгонки и отличается от прежней формулы для профиля без пропусков. Среднее ядро первого полубесконечного слоя дополнительно рассчитано при проекциях контактов, чтобы отделить правило усреднения от изменения электродов.

Три слоя означают фон, лёгкое и снова фон. Глобальная нижняя плоскость может отсутствовать в анатомии. Все интервалы сохраняют повторные пересечения, но тоже превращаются в плоские слои. Это диагностические приближения; большее число границ само по себе не доказывает точности. Центральный луч C01 пересекает один лёгочный интервал, поэтому варианты «первый центральный интервал» и «все центральные интервалы» здесь совпадают. Их одинаковые числа ожидаемы, а не являются независимым подтверждением точности.
