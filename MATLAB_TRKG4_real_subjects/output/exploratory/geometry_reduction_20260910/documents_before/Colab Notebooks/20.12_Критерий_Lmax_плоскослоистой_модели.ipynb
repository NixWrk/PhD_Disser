Сначала устанавливается пригодность выбранного семейства по ошибке базового
импеданса, целевых приращений, восстановленных параметров и вычислительной
стоимости. Для плоскослоистой модели дополнительно определяется допустимое
множество размеров и, когда это возможно, верхняя граница $L_{max}$.
Исторические 200 мм являются границей вычислительного поиска, а не результатом
проверки применимости.


## Что означает $L_{max}$

$L_{max}$ — не геометрический размер тела и не точка, где сигнал начинает
уменьшаться. Это граница применимости **сокращённой модели** относительно более
полной КТ/FEM-постановки.

Она зависит как минимум от:

- индивидуальной анатомии и положения матриц;
- состояния дыхания;
- размеров, формы и контакта электродов;
- электрических свойств тканей и частоты;
- выбранного выхода: абсолютный $Z$, дыхательное или пульсовое $\Delta Z$,
  восстановленные $\rho$ или другой параметр;
- заранее заданной допустимой ошибки.

Поэтому обозначение без аргументов допустимо только как сокращение. Полная
запись имеет вид

$$
L_{max}=L_{max}(G,P,D,S,Q,\varepsilon),
$$

где $Q$ — проверяемый выход, а $\varepsilon$ — допустимая ошибка.


## Почему одной оценки кривизны недостаточно

Тело не является поверхностью простой формы, а чувствительность тетраполярной
сборки распределена в объёме. Локальный радиус кривизны кожи не описывает:

- рёбра и межрёберные промежутки;
- неоднородную толщину мягких тканей $h(s)$;
- кривизну и конечную границу лёгкого;
- боковой уход тока и удалённые структуры;
- конечную площадь электродов и контакт;
- изменение всех перечисленных факторов вдоль крупной сборки.

Кривизна может быть одним из факторов или объясняющих признаков, но не
самостоятельным критерием $L_{max}$. Основной критерий должен строиться на
расхождении выходов согласованных плоской и полной 3D-моделей. Контролируемые
FEM-абляции затем могут отдельно показать вклад кривизны, рёбер, неоднородности
и конечной границы лёгкого.


## Контракт сравниваемых моделей

Для каждого размера $L$ вычисляются:

$$
Z^{3D}_L=F^{CT/FEM}_L(\Theta,G,P,D,S),
$$

$$
Z^{2L}_L=F^{flat}_L(\theta_{eff},h_{eff},D,S).
$$

Чтобы разность действительно характеризовала потерю при переходе к плоской
двуслойной модели, обе постановки должны иметь одинаковые единицы, частоту,
стимуляцию, порядок электродов, контакт и физиологический сценарий.

Правило получения эффективных $\theta_{eff}$ и $h_{eff}$ фиксируется **до**
проверки. Нельзя заново подбирать их отдельно для каждого $L$: такая подгонка
может скрыть структурную ошибку модели. Допустимы два честных протокола:

1. параметры заданы по независимой КТ/литературе/измерению и одинаковы для
   всех размеров;
2. параметры калибруются на заранее выбранной части размеров, а ошибка
   оценивается на отложенных размерах.

КТ/FEM здесь является более полной вычислительной референсной моделью, но не
физиологической истиной. Её собственная ошибка сетки, сегментации, свойств
тканей и электродов должна быть меньше допуска либо включена в отчёт.


### Продолжение двухтканного примера 20.10

В текущем сравнении сохраняются два сопротивления: лёгкое и все остальные
ткани, включая поддиафрагмальные. Первый и третий слои трёхслойного
приближения имеют одинаковые свойства. Материальная классификация должна
совпадать в сравниваемых задачах и не меняться одновременно с формой границы.

Исходный набор из [20.10](20.10_КТ_FEM_локализация_боковых_матриц_МГТУ.ipynb) использует номинальные точечные контакты
в аналитике и конечные контакты на кривой коже в FEM. Поэтому его расхождение
ещё не отделяет контактную ошибку от ошибки геометрического упрощения.
Продолжение начинается с согласования электродов и оценки сходимости референса.

Помимо двух- и трёхслойных формул рассматриваются параметрические кривые
поверхности, граничные элементы, 2,5D, модели пониженного порядка и табличные
аппроксимации. Ускорение решения при известной анатомии и исключение
индивидуальной КТ являются разными задачами. Полные определения методов
сохраняются в разделе 18 отчёта 20.10, без дублирования вычислительного ядра.


## Метрика и формальное определение

Для выхода $Q$ задаётся размерная ошибка

$$
E_Q^{abs}(L)=\left|Q^{2L}(L)-Q^{3D}(L)\right|
$$

и при необходимости нормированная ошибка

$$
E_Q^{norm}(L)=
\frac{E_Q^{abs}(L)}{Q_{scale}}.
$$

$Q_{scale}$ задаётся заранее из требуемой точности downstream-задачи,
разрешения измерения или независимого физиологического масштаба. Делить только
на $|Q^{3D}|$ опасно для дыхательных и пульсовых приращений, близких к нулю.

Если ошибка монотонно нарастает, индивидуальная граница определяется как

$$
L_{max}=\sup\{L:\ E_Q(l)\le\varepsilon
\ \text{для всех}\ l\le L\}.
$$

В реальной анатомии $E_Q(L)$ может быть немонотонной. Тогда нельзя
выбирать случайную последнюю точку ниже порога: следует сообщать допустимое
множество размеров и консервативную границу непрерывного допустимого префикса.
Для дискретно изготовленных сборок результат сначала остаётся дискретным;
интерполяция требует отдельного обоснования.

Порог $\varepsilon$ задаётся до просмотра итоговой кривой. Этот документ не
назначает численное значение: оно должно следовать из допустимого смещения
конечной оценки, а не подбираться так, чтобы сохранить желаемый размер.


## Какие выходы проверять раздельно

Одного сравнения абсолютного импеданса недостаточно. Нужны отдельные кривые:

1. абсолютный $Z_L$ в каждом состоянии;
2. дыхательное $\Delta Z_{resp,L}$ между вдохом и выдохом;
3. пульсовое $\Delta Z_{pulse,L}$ для явно заданных
   $\Delta\rho_{soft}$ и $\Delta\rho_{lung}$;
4. смещение параметров, которые обратная двуслойная задача восстанавливает из
   синтетических КТ/FEM-данных;
5. при использовании результата в коррекции ТТРКГ — смещение соответствующего
   поправочного фактора.

Размер может быть приемлем для абсолютного $Z$ и неприемлем для малого
$\Delta Z$. Поэтому у разных выходов могут быть разные $L_{max}$. Для выбора
рабочей пары серии `32` используется пересечение допустимых множеств по всем
обязательным выходам.

контактов и принятия критериев такое сравнение не устанавливает подтверждённый
индивидуальный $L_{max}$. Строгие и исследовательские статусы не смешиваются.


## Последовательность вычислительного эксперимента

1. На простой синтетической двуслойной геометрии проверить, что 3D FEM
   воспроизводит аналитическую модель в пределе одинаковых постановок.
2. Проверить сходимость FEM по сетке, размеру области и модели электродов.
3. Зафиксировать индивидуальную позу, свойства и правило получения
   $h_{eff}$; не переоптимизировать их скрыто при изменении $L$.
4. Для одной и той же сетки размеров вычислить полную 3D- и плоскую кривые.
5. Повторить сравнение для статического, дыхательного и пульсового сценариев.
6. В двухтканной постановке по одному изменять кривизну, неоднородность $h$
   и дальнюю границу лёгкого. Влияние отдельного костного материала проверять
   только в самостоятельном многоматериальном контрольном сценарии. Эти
   изменения объясняют источник расхождения, но не заменяют полный критерий.
7. Повторить по физически обоснованным диапазонам свойств, контакта и ошибки
   позы без необоснованных вероятностных распределений.
8. Повторить на второй индивидуальной анатомии; только затем обсуждать
   диапазонную или консервативную межсубъектную границу.

Последовательная регистрация сборок добавляет экспериментальную
неопределённость при сравнении с реальными данными, но не должна искусственно
встраиваться в чистое FEM-сравнение моделей как единый «шум».


Для сравнения разных семейств дополнительно измеряются стоимость подготовки,
время повторной оценки и память. Ошибка восстановленных сопротивлений
проверяется на синтетических данных с известными параметрами и на случаях,
не использованных при настройке. Повторная подгонка каждой модели по всем
девяти значениям не является независимой проверкой.

Итогом должен быть выбранный компромисс при заранее заданных допусках,
либо отрицательный результат о непригодности исследованных упрощений.
Сами допустимые ошибки в настоящем контракте численно не назначены.


## Что не является доказательством $L_{max}$

- `L_MAX_MM = 200` в `30.04` — только историческая заглушка. Любая оптимизация
  размеров, использующая этот верхний предел, условна до выполнения `20.12`.
- Спад или выброс реального сигнала на 140 мм сам по себе не доказывает
  нарушение плоскослоистости: его может вызвать систематика конкретной
  сборки, контакт, запись или прибор.
- Локальный радиус кривизны не учитывает объёмную анатомию.
- Хорошая подгонка плоской модели к тем же размерам, на которых параметры
  оценивались, не является независимой проверкой.
- Малый остаток $Z$ не гарантирует малую ошибку восстановленных $\rho$ или
  физиологических приращений.
- Код или сохранённый график без хешей входов, сеточного теста и заранее
  заданного порога не устанавливает $L_{max}$.

$L_{min}$ по измеримости/идентифицируемости и $L_{max}$ по ошибке сокращённой
модели — разные границы. Рабочий диапазон существует только при
$L_{min}<L_{max}$ для требуемой задачи.

## Критерии принятия

Значение или диапазон $L_{max}$ можно использовать дальше только если:

1. сравниваемые модели согласованы по физическим входам и единицам;
2. эффективные параметры не переобучаются на каждом проверяемом размере;
3. порог ошибки и целевой выход зарегистрированы заранее;
4. численная ошибка FEM существенно меньше принятого допуска либо её вклад
   явно учтён;
5. проверены абсолютный, дыхательный, пульсовый и downstream-выходы, которые
   реально используются далее;
6. показана устойчивость к обоснованным диапазонам свойств, контакта, позы и
   сегментации;
7. немонотонная ошибка представлена допустимым множеством, а не скрыта одним
   удобным числом;
8. индивидуальный результат не переносится на эксперимент 3 или других людей
   без отдельной проверки;
9. в серии `32` используется только принятый результат, а не заглушка 200 мм.

До выполнения этих условий статус остаётся `contract_only`.
