Это литературно мотивированная рабочая область, а не установленный физиологический интервал для данного человека или смеси всех фоновых тканей. Анизотропия и мнимая часть импеданса не учитываются.

Объединены прежние 16 внутренних сочетаний и 49 сочетаний сетки, охватывающей границы: всего **65 заданных пар**. Истинные сопротивления известны по условию вычислительного эксперимента. Вектор девяти импедансов получен прямым КТ/FEM-расчётом при каждом необходимом отношении сопротивлений; он не создаётся интерполятором обратной модели. Повторяющиеся отношения используют точное масштабирование по $\rho_1$.

Для каждой модели минимизируется один и тот же критерий:

$$ (\widehat\rho_1,\widehat\rho_2)=\underset{(\rho_1,\rho_2)\in B}{\operatorname{argmin}}\ \frac{1}{9}\sum_L[Z_L^{\mathrm{модель}}(\rho_1,\rho_2)-Z_L^{\mathrm{данные}}]^2, $$

где $\widehat\rho_1,\widehat\rho_2$ — оценки сопротивлений, Ом·м; $B$ — рабочая область; $L$ — один из девяти размеров; $Z_L^{\mathrm{модель}}$ — рассчитанный импеданс, Ом; $Z_L^{\mathrm{данные}}$ — синтетическое либо измеренное значение, Ом. Квадратный корень из минимального критерия обозначен RMSE. Равные веса в омах — расчётное допущение: ковариация ошибок неизвестна. Произвольный множитель или смещение шкалы не подгоняются.

Ограничения накладываются на оба сопротивления. Для точечных электродов при фиксированном отношении допустимый масштаб определяется аналитически, после чего ищутся минимумы по отношению. Для CEM с фиксированным контактом оба параметра оптимизируются численно: такое масштабирование в общем случае неприменимо. Плотность поиска проверяется отдельно. Попадание на границу сохраняется в результатах: оно может отражать несовпадение модели с данными, плохую различимость либо истинное граничное значение синтетического теста.

## 3. Интерпретация синтетического результата

Каждая модель восстанавливает оба параметра по одному референсному вектору. В 20.16 ошибки приведены относительно заданных сопротивлений, а не плоской формулы. Почти нулевая ошибка КТ-инверсии проверяет вычислительную цепочку; общая тетраэдральная сетка ограничивает независимость этой проверки.

Большая чувствительность к $\rho_2$ полезна, если размерные зависимости влияния двух тканей достаточно различаются. Число обусловленности в расчётных таблицах — отношение наибольшего сингулярного числа Якобиана по $\ln\rho_1,\ln\rho_2$ к наименьшему. Большое отношение означает, что некоторые сочетания относительных изменений параметров различимы гораздо хуже других. Это локальная характеристика при заданных весах и параметризации, не доверительный интервал.
