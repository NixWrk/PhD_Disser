# 31.05. Влияние допущений на восстановление удельных сопротивлений тканей

Цель исследования — определить, какие упрощения позволяют восстанавливать удельные сопротивления мягких тканей и лёгкого с требуемой точностью по небольшому числу боковых электродных сборок. Оцениваются три группы величин: базовые значения, их изменения при дыхании и пульсовые изменения в течение сердечного цикла. Удельное сопротивление обозначается $\rho$ и измеряется в Ом·м; измеряемый сборкой импеданс обозначается $Z$ и измеряется в омах.

Основная трудность обратной задачи состоит в том, что одинаковое изменение импеданса может быть вызвано разными причинами: изменением свойств тканей, толщины слоя или условий измерения. Поэтому малое расхождение между рассчитанным и измеренным импедансами само по себе не устанавливает точность найденных тканевых параметров. Здесь проверяются три конкретных механизма такой ошибки.

**Область полученных результатов.** Выполнены три вычислительных опыта с плоской двухслойной геометрией и идеальными точечными электродами. Они объясняют влияние отдельных допущений на восстановление параметров. Оптимальная модель, место измерения и минимальное число сборок для человека в этих опытах ещё не определены.

## 0. Исходные условия и принятые допущения

По решению автора от 11 сентября 2026 года индивидуальная компьютерная томография (КТ) считается доступной. Она задаёт анатомию выбранного места и позволяет определить расстояния от кожи до лёгкого. При кривых поверхностях это распределение расстояний, а не автоматически одна толщина. Способ замены распределения эффективной толщиной $h$ должен быть указан и проверен для выбранной геометрической модели. Такая замена пока не обосновывается одним наличием КТ.

Рассматриваются два варианта работы с толщиной: сохранить значение, определённое по КТ, постоянным либо оценивать его изменение по импедансам конкретного человека с опорой на состояние, представленное на КТ. Если изменение нельзя оценить устойчиво, используются принятые автором предельные сценарии: до 2 мм для глубокого вдоха–выдоха и до 1 мм для обычного дыхания. Эти пределы задают область проверки чувствительности; они не получены в настоящем исследовании из измерений толщины.

Ограничение задаётся разностью двух состояний или полным размахом за один дыхательный цикл:

$$|h_B-h_A|\leq d_h,\qquad \max_s h(s)-\min_s h(s)\leq d_h,$$

где $h_A,h_B$ — толщины в двух сравниваемых состояниях; $s$ — координата дыхательного цикла; $d_h$ — допустимый размах: 2 мм при глубоком и 1 мм при обычном дыхании. Для двух состояний используется первое неравенство, для траектории — второе. Интервал $\pm d_h$ вокруг среднего допускал бы вдвое больший размах и здесь не принимается. Оба направления изменения относительно состояния КТ проверяются как отдельные сценарии.

Дыхательное состояние КТ должно соответствовать исходному состоянию расчёта. В частности, ограничение размаха обычного дыхания не задаёт разность между его средней толщиной и толщиной на КТ, выполненной на глубоком вдохе. Эту разность необходимо определять отдельно либо задавать отдельным сценарием. Отказ от индивидуальной КТ сейчас не требуется.

В основной модели сохраняются две эффективные ткани: $\rho_1$ описывает нелёгочный фон, $\rho_2$ — лёгкое. Область под диафрагмой включается в фон. Представление разных нелёгочных тканей одним параметром является упрощением состава среды. Неизменность базовой $\rho_1$ при дыхании и отсутствие её пульсового изменения — два дополнительных допущения; автором они пока не приняты как доказанно допустимые.

## 1. Связь прямой модели с целью обратного расчёта

Прямая задача вычисляет импедансы по известным свойствам тканей, геометрии и условиям контакта. Обратная задача использует эти импедансы для оценки неизвестных свойств тканей. В вычислительном опыте истинные параметры заданы исследователем, поэтому ошибку восстановления можно вычислить непосредственно. У человека КТ даёт геометрию, но истинные $\rho_1$ и $\rho_2$ остаются неизвестными. Совпадение импедансов у человека имеет поэтому более ограниченный смысл, чем восстановление заранее заданных параметров в опыте.

В исследовании сохраняются два анатомических направления. [Уточнение геометрической модели при существующем монтаже](20.14_Две_проверки_моделей_с_литературными_сопротивлениями.ipynb) продолжает расчёты 20.13–20.14. [Выбор анатомически подходящего места](20.15_Анатомический_выбор_места_боковых_сборок.ipynb) рассматривается отдельно в 20.15. Оба направления должны оцениваться по ошибке восстановленных тканевых параметров, включая дыхательную и пульсовую составляющие.

Настоящий ноутбук связывает их с [анализом идентифицируемости](31.03_Идентифицируемость_двуслойной_задачи.ipynb), [разбором источников погрешности](31.04_Погрешности_бокового_эксперимента.md), выбором [пары при известной толщине](32.02_Выбор_пары_при_известной_h.ipynb) и [набора при неизвестной толщине](32.03_Выбор_набора_при_неизвестной_h.ipynb). Расчёты [базовых](33.01_Статическая_оценка_параметров_боковых_сборок.ipynb) и [пульсовых параметров](33.04_Пульсовые_изменения_удельного_сопротивления.ipynb) используют соответствующие ограничения модели.

В [совместной оценке вдоха и выдоха](33.07_Профиль_согласованности_удельного_сопротивления_лёгкого.ipynb) уже приняты общие $h$ и $\rho_1$, а $\rho_2$ различается между состояниями. Такая постановка уменьшает число неизвестных. Однако успешная подгонка не показывает, какой была бы ошибка при нарушении этих ограничений. Именно это обстоятельство определяет логику следующих опытов. Результаты и границы параметров прежнего расчёта здесь не изменяются.

## 2. Базовые, дыхательные и пульсовые параметры

Сначала необходимо разделить величины, точность которых проверяется. Базовое сопротивление относится к определённому дыхательному состоянию. Пульсовая составляющая описывает изменение в течение сердечного цикла при этом состоянии. Используется разложение:

$$\rho_j(s,\varphi)=\bar\rho_j(s)+\delta\rho_j^{\mathrm{п}}(s,\varphi),\qquad
\langle\delta\rho_j^{\mathrm{п}}(s,\varphi)\rangle_{\varphi}=0,$$

где $j=1,2$ — номер ткани; $s$ — дыхательное состояние или координата дыхательного цикла; $\varphi$ — фаза сердечного цикла; $\bar\rho_j$ — базовое удельное сопротивление; $\delta\rho_j^{\mathrm{п}}$ — знаковая пульсовая составляющая в Ом·м; угловые скобки обозначают выбранное усреднение по фазе. Нулевая средняя задаёт способ разделения составляющих. Для коротких переходных записей возможность такого усреднения требует отдельной проверки.

Дыхательное изменение определяется разностью базовых значений:

$$\Delta\rho_j^{\mathrm{д}}=\bar\rho_j(s_B)-\bar\rho_j(s_A),$$

где $s_A,s_B$ — два сравниваемых дыхательных состояния, например задержки после вдоха и выдоха; $\Delta\rho_j^{\mathrm{д}}$ — дыхательная разность в Ом·м. В обычном дыхании и базовый уровень, и пульсовая форма могут зависеть от дыхательной фазы.

| Оцениваемая величина | Значение для методики | Показатель точности |
|---|---|---|
| Базовые $\bar\rho_1,\bar\rho_2$ | Определяют исходное состояние тканей и чувствительность модели | Ошибка каждой ткани в Ом·м и процентах |
| Дыхательные $\Delta\rho_1^{\mathrm{д}},\Delta\rho_2^{\mathrm{д}}$ | Разделяют изменение двух тканей между состояниями | Ошибка разности и её знака |
| Пульсовые $\delta\rho_1^{\mathrm{п}},\delta\rho_2^{\mathrm{п}}$ | Нужны для оценки тканевых вкладов в ТТРКГ | Ошибка знаковой формы, амплитуды и временного положения |
| Тканевая поправка к ТТРКГ | Связывает боковые измерения с последующей методикой | Ошибка переносимого сигнала при отдельно проверенном операторе переноса |

У малой разности относительная ошибка может быть большой даже при небольшой ошибке исходных уровней. Вблизи нуля следует использовать абсолютную ошибку или заранее заданную опорную амплитуду. Разность максимума и минимума сигнала не сохраняет его знак и фазу. Электрокардиограмма задаёт временную привязку, но сама по себе не разделяет вклад мягких тканей, лёгкого и сердца.

## 3. Влияние геометрии и контактов на оцениваемые параметры

Измерение зависит не только от удельных сопротивлений. Общая запись модели имеет вид:

$$Z_i(s,\varphi)=F_i\!\left(\rho_1(s,\varphi),\rho_2(s,\varphi),h(s,\varphi),
g(s,\varphi),c_i(s,\varphi)\right)+e_i(s,\varphi),$$

где $i$ — номер сборки; $F_i$ — оператор прямого расчёта; $h$ — толщина в упрощённой модели; $g$ — остальные геометрические параметры, включая форму границ, положение и ориентацию сборки; $c_i$ — параметры контактов; $e_i$ — расхождение модели с наблюдением. Для этого расхождения здесь не задаются независимость, гауссовское распределение или общая дисперсия. В полном описании КТ толщина выводится из поверхностей; её нельзя повторно вводить как независимую деформацию той же геометрии.

Произвольные изменения всех перечисленных величин нельзя восстановить по двум импедансам. На этапе разработки часть величин задаётся по внешним данным или сценариям. Затем оценивается ошибка, которую вносят их фиксация и неточность.

Для малых пульсовых изменений около заданного базового состояния используется линейное приближение:

$$\delta\mathbf Z^{\mathrm{п}}\simeq
J_{\rho}(s)\delta\boldsymbol\rho^{\mathrm{п}}
+\mathbf J_h(s)\delta h^{\mathrm{п}}
+J_g(s)\delta\mathbf g^{\mathrm{п}}
+J_c(s)\delta\mathbf c^{\mathrm{п}},$$

где $\delta\mathbf Z^{\mathrm{п}}$ — вектор пульсовых изменений импедансов; $\delta\boldsymbol\rho^{\mathrm{п}}$ — изменения двух удельных сопротивлений; $\delta h^{\mathrm{п}},\delta\mathbf g^{\mathrm{п}},\delta\mathbf c^{\mathrm{п}}$ — изменения толщины, прочей геометрии и контактов; матрицы $J$ содержат соответствующие производные прямого оператора. Матрицу производных также называют якобианом: она показывает, как малое изменение каждого параметра меняет измерения. Производные вычисляются для текущего дыхательного состояния. Ошибка базовых параметров меняет эти производные и может передаваться в пульсовую оценку.

Переход между глубоким вдохом и выдохом сначала следует рассчитывать как разность двух полных нелинейных решений. Применимость одного якобиана ко всему переходу требует проверки. Кроме того, усреднение нелинейного отклика не обязано давать отклик при усреднённых параметрах. Возможная ошибка такого усреднения входит в проверку способа выделения базового уровня.

## 4. Число неизвестных при разных предположениях о толщине

При двух дыхательных состояниях и известной толщине в каждом из них необходимо определить четыре базовых сопротивления: по два для состояния A и B. Две сборки дают четыре импеданса. Если толщина известна только в состоянии КТ, а во втором состоянии также подбирается, неизвестных становится пять при тех же четырёх наблюдениях. В общем случае такой информации недостаточно. Дополнительные размеры или обоснованная связь между параметрами могут устранить этот недостаток, но устойчивость решения затем проверяется отдельно.

| Постановка | Число неизвестных | Содержание ограничения |
|---|---:|---|
| Отдельные $\rho_1,\rho_2,h$ в двух состояниях | 6 | Общая постановка без использования толщины по КТ |
| $h_A$ известна по КТ, $h_B$ подбирается, сопротивления отдельные | 5 | Текущий вариант восстановления изменения толщины |
| Общая $\rho_1$, отдельные $\rho_2,h$ | 5 | Базовое сопротивление фона не меняется при дыхании |
| Общая неизвестная $h$, отдельные сопротивления | 5 | Толщина постоянна, но её значение ещё определяется |
| Общие неизвестные $\rho_1,h$, отдельные $\rho_2$ | 4 | По числу параметров соответствует совместной схеме 33.07 |
| Обе толщины известны, сопротивления отдельные | 4 | Геометрия задана внешними данными |
| Обе толщины известны, базовая $\rho_1$ общая | 3 | Дополнительно принято постоянство одного тканевого параметра |

Строки без заданной по КТ толщины приведены для сравнения числа неизвестных; они не отменяют её доступность в текущей задаче. Подсчёт параметров даёт необходимую предварительную проверку. Даже при достаточном числе наблюдений разные сочетания параметров могут давать почти одинаковые импедансы — тогда решение остаётся неустойчивым.

Известные $h_A$ и изменение $h_B-h_A$ задают обе толщины. Если известна только разность, исходная толщина остаётся неизвестной. Аналогично, заданная форма закона $h(s)=h_0+\alpha q(s)$ не исключает неизвестные $h_0$ и $\alpha$: здесь $q(s)$ — выбранный показатель дыхания, $h_0$ — исходный уровень, $\alpha$ — коэффициент связи. Показатель, оценённый из тех же импедансов, не является независимым внешним измерением.

Таким образом, фиксация параметра уменьшает размер обратной задачи, но не устраняет погрешность его заданного значения. В частности, постоянство базовой $\rho_1$ между вдохом и выдохом не означает отсутствия пульсовой $\delta\rho_1^{\mathrm{п}}$ внутри каждого состояния. Эти ограничения проверяются раздельно.

## 5. Логика вычислительных опытов и общие условия сравнения

Чтобы определить влияние одного допущения, синтетические импедансы рассчитываются с известным ненулевым нарушением этого допущения. При восстановлении параметров нарушение не учитывается. Разность между заданными и восстановленными параметрами показывает последствия такого сокращения в выбранной модели. Если один и тот же эффект исключить и при генерации, и при восстановлении, его влияние останется непроверенным.

Три опыта образуют последовательность проверок, но численные оценки одного опыта не подставляются в следующий. Перед каждым новым опытом исходные условия задаются заново. Благодаря этому ошибки разных допущений пока не смешиваются.

| Опыт и рисунок | Проверяемое сокращение | Что задаётся точно для выделения его влияния | Величина, по которой оценивается результат |
|---|---|---|---|
| 1. Толщина, рисунок 1 | Значение $h$ фиксируется, хотя истинная толщина отличается | Два истинных сопротивления; прочая геометрия | Ошибка каждого базового сопротивления |
| 2. Дыхательное изменение, рисунок 2 | Базовая $\rho_1$ считается общей для двух состояний | Толщина обоих состояний | Ошибка дыхательной разности $\Delta\rho_2$ |
| 3. Пульсовое изменение, рисунок 3 | Пульсовая составляющая $\rho_1$ исключается | Базовые сопротивления и толщина | Ошибка знаковой пульсовой составляющей лёгкого |

**Общая физическая модель.** Во всех опытах используются две идеальные точечные сборки 50 и 140 мм на плоской поверхности. Для сборки с размером $L$ токовые точки расположены в $\pm L/2$, измерительные — в $\pm L/4$ вдоль одной прямой. Источник тока находится слева, сток — справа; измеряется потенциал левой внутренней точки минус потенциал правой. Ток 1 А используется только для нормировки расчётного импеданса и не обозначает ток прибора. Площадь электродов и контактный импеданс в модели отсутствуют.

Первый слой имеет постоянную толщину внутри каждого расчётного состояния, второй занимает нижнее полупространство. Свойства каждого слоя однородны, изотропны и описываются действительным удельным сопротивлением. Поэтому опыты не учитывают пространственное распределение толщины, кривизну, край лёгкого, диафрагму, реактивную составляющую или анизотропию. Эта идеализация позволяет выделить ошибку самого ограничения параметров. Она не устанавливает численную величину ошибки для КТ-геометрии.

**Исходные значения и границы.** Опорные сопротивления равны 4 и 16 Ом·м, толщина — 30 мм. Это заданная рабочая точка, а не результат определения параметров человека. Пара 50/140 мм используется для демонстрации и не объявляется оптимальной. В нелинейной подгонке сохранена [литературная рабочая область](20.14_Две_проверки_моделей_с_литературными_сопротивлениями.ipynb): $\rho_1$ приблизительно 1,377–10 Ом·м и $\rho_2$ приблизительно 9,741–24,074 Ом·м. Область ограничивает поиск, но не является популяционным нормативом эффективных тканей при 50 кГц. В показанных решениях её границы не достигаются.

Синтетические импедансы не содержат добавленного измерительного шума. В первых двух опытах минимизируется сумма квадратов относительных остатков: разность рассчитанного и заданного импедансов делится на заданный импеданс каждого наблюдения. Равные веса выбраны для этих опытов и не выведены из дисперсии прибора. Подгонка выполнена из трёх начальных точек с проверкой согласия результатов. В таблицах RMS означает среднеквадратический остаток; его единицы указаны в заголовке. Значение False в столбце «На границе» означает, что решение не достигло границ поиска.

Принцип компенсации можно выразить через локальную оценку смещения:

$$\delta\widehat{\boldsymbol\theta}\simeq
(J_\theta^{\mathsf T}WJ_\theta)^{-1}J_\theta^{\mathsf T}WJ_\eta\,\delta\boldsymbol\eta,$$

где $\boldsymbol\theta$ — восстанавливаемые параметры; $\boldsymbol\eta$ — ошибочно фиксированные параметры; $J_\theta,J_\eta$ — производные прямого оператора по этим группам; $W$ — положительно определённая матрица весов; $\delta\boldsymbol\eta$ — истинное значение минус заданное; $\delta\widehat{\boldsymbol\theta}$ — вызванное этим смещение оценки. Формула применима при малых возмущениях, независимых столбцах $J_\theta$ и неактивных границах. Она показывает, какая часть неучтённого изменения может быть объяснена изменением оцениваемых параметров. Конечные ошибки в первых двух опытах получены нелинейной подгонкой.

## 6. Ошибка сопротивлений при неверно заданной толщине

Первый опыт относится к принятому варианту постоянной толщины. Его задача — установить, какое ложное изменение сопротивлений возникает, если толщина фактически изменилась, а в обратной модели сохранилось значение по КТ.

В исходном состоянии толщина равна 30 мм. Для каждого отдельного сценария синтетические импедансы второго состояния рассчитываются при толщине от 28 до 32 мм. Истинные сопротивления в обоих состояниях сохраняются равными 4 и 16 Ом·м. Затем по двум импедансам второго состояния подбираются оба сопротивления при ошибочно фиксированной толщине 30 мм.

Каждая строка таблицы описывает отдельное изменение относительно 30 мм. Например, 28 и 32 мм относятся к разным сценариям; переход между ними в одном цикле не моделируется. Разности по модулю до 1 мм соответствуют принятому пределу обычного дыхания, до 2 мм — глубокого. Постоянные истинные сопротивления выбраны специально, чтобы вся ошибка их восстановления была вызвана только неверной толщиной. Полный физиологический переход вдох–выдох, при котором свойства тканей также могут изменяться, здесь не воспроизводится.

```python
from pathlib import Path
import json, sys, hashlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import least_squares
from IPython.display import display

nbdir = Path.cwd()
if not (nbdir/'two_layer_model.py').exists(): nbdir = nbdir/'Colab Notebooks'
sys.path.insert(0,str(nbdir))
from two_layer_model import forward_curve
SIZES = np.array([.05,.14])
BOUNDS = np.array([[1.3774104683195594,10.0],[9.741271489527174,24.074073806584366]])
RHO = np.array([4.,16.]); H = .03

def fit(fun, lower, upper, starts):
    fits = [least_squares(fun,np.log(x),bounds=(np.log(lower),np.log(upper)),
                         xtol=1e-12,ftol=1e-12,gtol=1e-12,max_nfev=1000) for x in starts]
    assert all(x.success for x in fits)
    best = min(fits,key=lambda x: np.dot(x.fun,x.fun))
    estimates = np.array([np.exp(x.x) for x in fits])
    assert np.max(np.ptp(estimates,axis=0)/np.exp(best.x)) < 1e-5
    estimate = np.exp(best.x)
    at_bound = bool(np.any(np.isclose(estimate,lower,rtol=1e-5)) or np.any(np.isclose(estimate,upper,rtol=1e-5)))
    return estimate,best.fun,at_bound

thickness_rows = []
for dh in [-2.,-1.5,-1.,-.5,-.25,-.1,0.,.1,.25,.5,1.,1.5,2.]:
    target,_ = forward_curve(SIZES,*RHO,H+dh/1000)
    fun = lambda x: (forward_curve(SIZES,*np.exp(x),H)[0]-target)/target
    est,res,bound = fit(fun,BOUNDS[:,0],BOUNDS[:,1],[[4,16],[3,12],[6,22]])
    thickness_rows.append([dh,*est,*(100*(est/RHO-1)),np.sqrt(np.mean((target*res)**2)),bound])
thickness = pd.DataFrame(thickness_rows,columns=['h истинная − h заданная, мм','ρ₁ оценка, Ом·м','ρ₂ оценка, Ом·м','Ошибка ρ₁, %','Ошибка ρ₂, %','RMS остатка, Ом','На границе'])
assert thickness['RMS остатка, Ом'].max()<1e-7
assert not thickness['На границе'].any()
shown = thickness.round(6)
shown['RMS остатка, Ом'] = thickness['RMS остатка, Ом'].map(lambda v: f'{v:.2e}')
display(shown)
plt.rcParams.update({'font.family':'DejaVu Sans','font.size':11,'axes.spines.top':False,'axes.spines.right':False})
fig,ax = plt.subplots(figsize=(9,4),constrained_layout=True)
ax.plot(thickness.iloc[:,0],thickness['Ошибка ρ₁, %'],'o-',label='ρ₁: мягкие ткани')
ax.plot(thickness.iloc[:,0],thickness['Ошибка ρ₂, %'],'s-',label='ρ₂: лёгкое')
ax.axvspan(-2,2,color='#e2ba72',alpha=.13,label='Отдельные сценарии глубокого дыхания: |Δh| ≤ 2 мм')
ax.axvspan(-1,1,color='#64a5b3',alpha=.15,label='Отдельные сценарии обычного дыхания: |Δh| ≤ 1 мм')
ax.axhline(0,color='gray',lw=.8); ax.grid(alpha=.2); ax.legend(fontsize=9)
ax.set(xlabel='Истинная толщина минус фиксированная, мм',ylabel='Ошибка восстановленного сопротивления, %',title='Оба импеданса воспроизведены; сопротивления могут быть смещены')
plt.show()
```

**Рисунок 1. Влияние ошибки заданной толщины на базовые удельные сопротивления.** По горизонтали отложена истинная толщина минус фиксированные 30 мм. Отрицательные значения означают, что слой в прямом расчёте тоньше принятого при восстановлении. По вертикали показана знаковая относительная ошибка каждого сопротивления: положительная соответствует завышению, отрицательная — занижению. Цветные области обозначают отдельные сценарии в принятых пределах 1 и 2 мм. Промежутки между рассчитанными точками соединены линиями для чтения зависимости.

При истинной толщине 29,5 мм сопротивление лёгкого завышено на 3,68%, при 30,5 мм — занижено на 3,43%. Для отдельных изменений −1 и +1 мм ошибки составляют +7,62 и −6,64%; для −2 и +2 мм — +16,44 и −12,45%. Ошибка $\rho_1$ в этой рабочей точке значительно меньше. При этом среднеквадратический остаток двух импедансов не превышает $4{,}74\cdot10^{-14}$ Ом. Это точность численного совпадения в идеальном опыте, а не точность измерительного прибора.

**Вывод из первого опыта.** Неверная толщина может быть почти полностью компенсирована изменением найденных сопротивлений. Поэтому даже практически точное совпадение двух импедансов не исключает заметной ошибки $\rho_2$. В выбранном примере ограничение изменения толщины величиной 1–2 мм само по себе ещё не делает её фиксацию безвредной. Малую ошибку $\rho_1$ нельзя переносить на другие толщины, сопротивления или размеры сборок.

Полученные проценты относятся к одной плоской геометрии, одной паре и одному сочетанию сопротивлений без шума. Они не задают универсальной погрешности для человека. Неравномерная толщина и кривизна могут изменить саму зависимость. Кроме того, пока не проверено, как эта ошибка базовых оценок передаётся в пульсовые параметры.

Ошибка толщины выявлена, но её устранение ещё не гарантирует точного восстановления дыхательных изменений. В следующем опыте толщина задаётся точно, чтобы проверить уже другое сокращение: допустимо ли считать базовое сопротивление мягких тканей одинаковым в двух состояниях. Оценки из рисунка 1 в этот опыт не переносятся.

## 7. Ошибка дыхательной разности при общей базовой ρ₁

В состоянии A задаются $\rho_{1,A}=4$ и $\rho_{2,A}=16$ Ом·м. В состоянии B сопротивление лёгкого равно 12 Ом·м, а сопротивление фона отличается от 4 Ом·м на величину от −2 до +2%. Толщина обоих состояний точно известна и равна 30 мм. Диапазон изменения фона выбран для проверки чувствительности и не является установленной физиологической изменчивостью.

По четырём синтетическим импедансам — две сборки в двух состояниях — подбираются одно общее сопротивление фона и два сопротивления лёгкого. Целевой результат — дыхательная разность лёгочного сопротивления B минус A. Её истинное значение во всех сценариях равно −4 Ом·м. Так проверяется, насколько неверное ограничение на одну ткань искажает оценённое изменение другой.

Подгонка использует равные веса относительных остатков. В прежнем совместном расчёте 33.07 равные веса относятся к остаткам в омах. При несовместимом ограничении веса влияют на то, как распределяется ошибка между наблюдениями; поэтому данный опыт не является пересчётом результатов 33.07. Постоянная и точно известная толщина позволяет отнести найденное смещение к ограничению на $\rho_1$ в пределах выбранной модели.

```python
breath_rows = []
for change in [-2.,-1.,-.5,0.,.5,1.,2.]:
    target = np.array([forward_curve(SIZES,4,16,H)[0],forward_curve(SIZES,4*(1+change/100),12,H)[0]])
    def objective(x):
        r = np.exp(x)
        prediction = np.array([forward_curve(SIZES,r[0],r[1],H)[0],forward_curve(SIZES,r[0],r[2],H)[0]])
        return ((prediction-target)/target).ravel()
    lower = BOUNDS[[0,1,1],0]; upper = BOUNDS[[0,1,1],1]
    est,res,bound = fit(objective,lower,upper,[[4,16,12],[3,20,14],[6,12,20]])
    delta = est[2]-est[1]
    breath_rows.append([change,*est,delta,100*abs(delta+4)/4,100*np.sqrt(np.mean(res**2)),bound])
breath = pd.DataFrame(breath_rows,columns=['Истинное изменение ρ₁, %','Общая ρ₁, Ом·м','ρ₂ A, Ом·м','ρ₂ B, Ом·м','Δρ₂ оценка, Ом·м','Ошибка Δρ₂ относительно |−4|, %','RMS относительного остатка, %','На границе'])
assert not breath['На границе'].any()
zero = breath.loc[breath.iloc[:,0]==0].iloc[0]
assert zero['Ошибка Δρ₂ относительно |−4|, %'] < 1e-5
display(breath.round(5))
fig,ax = plt.subplots(figsize=(9,4),constrained_layout=True)
ax.plot(breath.iloc[:,0],breath['Ошибка Δρ₂ относительно |−4|, %'],'o-',color='#b64839')
ax.set(xlabel='Неучтённое изменение базовой ρ₁ между состояниями, %',ylabel='Модуль ошибки дыхательной Δρ₂, %',title='Цена ограничения: одно фоновое сопротивление для двух состояний')
ax.grid(alpha=.2);plt.show()
```

**Рисунок 2. Влияние неучтённого изменения базовой ρ₁ на дыхательную разность лёгкого.** Горизонтальная ось показывает истинное относительное изменение фона в состоянии B по сравнению с A. На вертикальной оси отложен модуль ошибки восстановленной $\Delta\rho_2$, делённый на модуль истинной разности 4 Ом·м и выраженный в процентах. В отличие от рисунка 1, эта ошибка не сохраняет знак; направление смещения определяется по оценённой разности в таблице.

При истинном увеличении $\rho_1$ на 1% ограниченная модель даёт лёгочные сопротивления 15,668 и 12,216 Ом·м. Их разность составляет −3,452 вместо −4 Ом·м: снижение сопротивления лёгкого недооценено, а относительная ошибка разности равна 13,69%. Среднеквадратический относительный остаток импедансов при этом составляет около 0,30%. При нулевом изменении фона ограничение соответствует заданной истине, и ошибка разности практически исчезает.

**Вывод из второго опыта.** Даже при точно известной толщине небольшое неучтённое изменение базового сопротивления мягких тканей может заметно исказить дыхательное изменение лёгкого. Ошибки двух лёгочных уровней здесь меньше ошибки их разности: при вычитании смещения уровней изменяют результат относительно сравнительно небольшой величины 4 Ом·м. Малый остаток импедансов вновь оказывается недостаточным критерием точности целевого параметра.

Опыт не устанавливает, меняется ли $\rho_1$ у человека на 1% или на другую величину. Он определяет последствие заданного нарушения гипотезы. Вывод зависит от выбранных сопротивлений, пары и весов, а ошибка толщины и измерительный шум в нём исключены. Проценты рисунков 1 и 2 нельзя непосредственно сравнивать как степень вреда двух допущений: они относятся к разным целевым величинам и разным нарушениям.

Далее необходимо рассмотреть более быстрые изменения внутри сердечного цикла. Постоянство среднего сопротивления между вдохом и выдохом не означает отсутствия пульсового изменения около этого среднего. Поэтому результат рисунка 2 не позволяет исключить пульсовой вклад мягких тканей. Для его отдельной проверки в третьем опыте заново задаются точные базовые параметры; смещённые оценки второго опыта не используются.

## 8. Ошибка лёгочной пульсовой составляющей при исключении фоновой

Базовые сопротивления снова равны 4 и 16 Ом·м, толщина — 30 мм; все три значения считаются точно известными. Геометрия и контакты при пульсе неизменны. Истинный малый сигнал содержит изменения обеих тканей, а сокращённая обратная модель разрешает меняться только сопротивлению лёгкого. Такая постановка выделяет ошибку исключения одного пульсового источника при отсутствии ошибки базового состояния.

Пульсовые изменения задаются общей синусоидальной временной функцией. Коэффициенты $q_1,q_2$ равны отношению знакового пульсового изменения к базовому сопротивлению соответствующей ткани. Для лёгкого $q_2=0{,}1\%$; фоновый коэффициент изменяется от −0,05 до +0,05%. Отрицательный коэффициент означает противоположную фазу. Эти малые величины и форма сигнала выбраны для вычислительного опыта и не описывают измеренные физиологические амплитуды.

Для анализа используются относительные изменения импедансов и безразмерная матрица чувствительности:

$$A_{ij}=\frac{\rho_j}{Z_i}\frac{\partial Z_i}{\partial\rho_j},\qquad
\widehat q_2-q_2=\frac{\mathbf a_2^{\mathsf T}\mathbf a_1}{\mathbf a_2^{\mathsf T}\mathbf a_2}\,q_1,$$

где $i$ — номер сборки; $j=1,2$ — ткань; $Z_i,\rho_j$ — базовые импедансы и сопротивления; $A_{ij}$ — относительная чувствительность; $\mathbf a_1,\mathbf a_2$ — столбцы матрицы $A$; $q_1,q_2$ — истинные знаковые относительные коэффициенты; $\widehat q_2$ — оценка лёгочного коэффициента при принятом $q_1=0$. Второе выражение получено для линейной модели с равными весами относительных изменений. Оно показывает, какую часть фонового сигнала сокращённая модель объясняет изменением лёгкого.

При сохранении обоих пульсовых параметров линейный расчёт восстанавливает заданные коэффициенты. Затем один параметр исключается и оценивается возникающая ошибка. Дополнительно малое линейное изменение импедансов сопоставляется с полным нелинейным расчётом в фазе максимального положительного значения общей синусоиды для каждого сценария.

```python
z,j = forward_curve(SIZES,*RHO,H)
A = j[:,:2]*RHO/z[:,None]
assert np.allclose(A.sum(axis=1),1,rtol=1e-10)
q2 = .001
alias = np.dot(A[:,1],A[:,0])/np.dot(A[:,1],A[:,1])
pulse_rows=[]; relative_linearization_errors=[]
for q1 in [-.0005,-.0001,0.,.0001,.0005]:
    signal=A@np.array([q1,q2])
    both=np.linalg.solve(A,signal)
    assert np.allclose(both,[q1,q2],atol=1e-14)
    reduced=np.dot(A[:,1],signal)/np.dot(A[:,1],A[:,1])
    exact=(forward_curve(SIZES,*(RHO*(1+np.array([q1,q2]))),H)[0]-z)/z
    relative_linearization_errors.append(np.linalg.norm(exact-signal)/np.linalg.norm(exact))
    pulse_rows.append([100*q1,100*q2,100*reduced,100*(reduced/q2-1)])
assert max(relative_linearization_errors)<.01
pulse=pd.DataFrame(pulse_rows,columns=['Истинная q₁, %','Истинная q₂, %','q₂ при исключении q₁, %','Знаковая ошибка q₂, %'])
display(pulse.round(5))
print(f'Коэффициент переноса относительной фоновой компоненты в лёгочную: {alias:.6f}.')
print(f'Максимальное относительное расхождение линейного и полного изменения сигнала в проверенных фазовых точках: {max(relative_linearization_errors):.3e}.')
fig,ax=plt.subplots(figsize=(9,4),constrained_layout=True)
ax.plot(pulse.iloc[:,0],pulse.iloc[:,2],'o-',label='Восстановлено при q₁ = 0')
ax.axhline(100*q2,color='#b64839',label='Истинная лёгочная компонента');ax.axhline(0,color='gray',lw=.8)
ax.set(xlabel='Неучтённая относительная фоновая компонента q₁, %',ylabel='Знаковая лёгочная компонента q₂, %',title='Малый тканевый пульс может заметно изменить оценку другой ткани')
ax.legend();ax.grid(alpha=.2);plt.show()
out = nbdir.parent/'MATLAB_TRKG4_real_subjects/output/exploratory/inverse_assumptions_20260911'
out.mkdir(parents=True,exist_ok=True)
for name,frame in [('thickness',thickness),('common_rho1',breath),('pulse_omission',pulse)]:
    frame.to_csv(out/(name+'.csv'),index=False)
checks={'thickness_max_residual_ohm':float(thickness['RMS остатка, Ом'].max()),
        'all_fits_interior':not bool(thickness['На границе'].any() or breath['На границе'].any()),
        'pulse_alias_gain':float(alias),'max_pulse_linearization_relative_error':float(max(relative_linearization_errors)),
        'new_fem_solves':0,'new_experimental_fits':0,
        'author_excursion_bounds_mm':{'deep':2,'ordinary':1},'three_start_agreement_rtol':1e-5}
_ = (out/'checks.json').write_text(json.dumps(checks,ensure_ascii=False,indent=2)+chr(10),encoding='utf-8')
```

**Рисунок 3. Искажение лёгочной пульсовой составляющей при неучтённом фоне.** По горизонтали показан истинный фоновый коэффициент $q_1$, по вертикали — лёгочный коэффициент в процентах от базового сопротивления лёгкого. Горизонтальная линия соответствует истинному $q_2=0{,}1\%$, наклонная — его оценке при исключении фонового изменения. Подпись «Восстановлено при $q_1=0$» относится к ограничению обратной модели; истинный $q_1$ задан горизонтальной осью.

Коэффициент переноса фонового изменения в лёгочную оценку равен 4,2543 для этой пары, рабочей точки и весов. При истинном $q_1=0{,}01\%$ оценка лёгочного коэффициента составляет около 0,14254% вместо 0,1%; относительная ошибка равна +42,54%. При $q_1=-0{,}05\%$ оценка становится отрицательной, хотя заданный лёгочный коэффициент положителен. Следовательно, исключение фона способно исказить не только величину, но и знак результата.

**Вывод из третьего опыта.** Малость пульсового изменения мягких тканей относительно их базового сопротивления не доказывает, что этим изменением можно пренебречь при оценке лёгкого. Его необходимо сравнивать с допустимой ошибкой лёгочной составляющей с учётом чувствительности сборок. Коэффициент 4,2543 характеризует математическое смешение двух тканевых вкладов; он не является долей кровенаполнения.

Максимальное относительное расхождение линейного и полного изменений в проверенных фазовых точках составляет около 0,114%. Это подтверждает малость ошибки линейного приближения в этих точках и позволяет отделить её от большого смещения при исключении фона. Проверка не охватывает всю временную обработку, произвольные формы и задержки пульса. Точная базовая точка и неподвижная геометрия также приняты специально; влияние ошибок базовых параметров на пульсовые оценки остаётся непроверенным.

В результате трёх опытов выделены разные механизмы ошибки: компенсация неверной толщины базовыми сопротивлениями, искажение дыхательной разности при общей $\rho_1$ и смешение пульсовых вкладов. Следующая задача — проверить их совместное действие в условиях конкретного дыхательного протокола. Для этого сначала требуется разделить глубокое и обычное дыхание, поскольку набор доступных состояний и возможность усреднения в них различаются.

## 9. Применимость результатов к дыхательным протоколам

Первые два опыта сравнивают отдельные состояния, третий относится к малым изменениям около одного состояния. Этого недостаточно для описания полной непрерывной записи: при дыхании могут одновременно меняться базовые параметры, геометрия и чувствительность к пульсу. Поэтому перенос результатов проверяется раздельно для следующих условий.

| Протокол | Рабочее описание | Необходимая дополнительная проверка |
|---|---|---|
| Задержки после глубокого вдоха и выдоха | Два базовых состояния и пульсовые формы внутри каждого | Соответствие состоянию КТ, стабильность задержек, нелинейная разность и отдельные якобианы |
| Непрерывное глубокое дыхание | Переход между крайними состояниями | Движение, контактные изменения и пригодность закона изменения толщины |
| Обычное дыхание | Базовые уровни и пульсовые формы, зависящие от дыхательной фазы | Воспроизводимость циклов, способ разделения временных составляющих и применимость ограничений, принятых для задержек |

Задержка дыхания не устанавливает полной неподвижности или постоянства контактов. Фильтрация и привязка к ЭКГ помогают выделять временные составляющие, но не добавляют независимых измерений их пространственного происхождения. Две произвольные тканевые траектории, изменение толщины и контактные изменения нельзя свободно разделить по двум каналам без дополнительных связей или внешней информации.

При последовательной записи разных размеров общий набор импедансов относится к разным моментам времени. Сопоставимость состояния становится допущением, которое требует проверки воспроизводимости. Одновременные или быстро переключаемые измерения уменьшают этот временной разрыв, но требуют проверки межканального влияния. Выбор протокола должен учитывать обе группы ошибок.

## 10. Последующие проверки и основания для сокращения модели

Полученные примеры показывают необходимость проверки допущений по ошибке целевых параметров. Для выбора методики нужны также величины допустимой ошибки: отдельно для базовых уровней, дыхательных разностей, пульсовой формы и тканевой поправки к ТТРКГ. Пока эти требования численно не заданы, следует показывать зависимости ошибок от условий, без признания модели достаточно точной по произвольному порогу.

Следующий вычислительный этап должен охватить диапазоны сопротивлений, толщин и размеров сборок, а затем реальные КТ-геометрии. Сначала проверяется обратное решение с сохранённым исследуемым эффектом. Если даже эта постановка не разделяет параметры, сокращение неизвестных не считается доказательством их точного определения. Далее эффекты исключаются по одному и в сочетаниях; ошибка сравнивается с известной синтетической истиной.

| Допущение для проверки | Что варьировать в задаче с известной истиной | Какое последствие оценивать |
|---|---|---|
| Постоянная либо точно заданная $h$ | Разность толщин, ошибка исходного уровня и форма границ | Смещение базовых параметров и передача ошибки в пульсовую оценку |
| Известный закон изменения $h$ | Ошибка его уровня, коэффициентов и временной формы | Ошибка восстановленных тканевых траекторий |
| Общая базовая $\rho_1$ | Её изменение между дыхательными состояниями | Ошибка обоих лёгочных уровней и их разности |
| Нулевая пульсовая составляющая фона | Её амплитуда, форма и задержка | Ошибка величины, формы и знака лёгочного пульса |
| Постоянные геометрия и контакт при пульсе | Малые перемещения и контактные возмущения | Ложное тканевое изменение |
| Один пульсовой якобиан для всего дыхания | Базовые состояния и геометрия дыхательного цикла | Ошибка пульсовой амплитуды в зависимости от дыхательной фазы |
| Достаточная точность плоской модели | Кривизна, распределение толщины и конечный объём лёгкого | Ошибки всех восстанавливаемых параметров |
| Малый вклад сердца и других органов | Органные чувствительности, допустимые контрасты и движение | Примесь к базовым, дыхательным и пульсовым тканевым оценкам |

Совместная проверка необходима, потому что ошибки могут компенсироваться в одном состоянии и усиливать друг друга в другом. Отдельно сравниваются пульсовые оценки при точно заданных и ошибочно восстановленных базовых параметрах. Так определяется влияние первого этапа инверсии на последующий, которое в трёх текущих опытах намеренно разделено.

Кандидаты геометрической модели, места и размеров следует сравнивать по одной системе требований. Сначала используется полный доступный набор размеров, затем сборки исключаются по одной с контролем ошибки и устойчивости. Ровно две сборки остаются целевым сокращением, а не заранее доказанным минимумом. Геометрический отбор мест из 20.15 и проверка упрощений из 20.13–20.14 включаются в это сравнение.

Для независимости проверки правило выбора фиксируется на одной группе людей и оценивается без перенастройки на другой. Фазы и повторные серии одного человека должны оставаться в одной группе. КТ обеспечивает разнообразие геометрий, но не задаёт истинные электрические свойства; конечноэлементный референс также требует проверки сетки, электродов, взаимности и производных. Отдельная физическая проверка возможна на фантомах с известными свойствами. Повторные измерения и новые сборки у добровольцев проверяют воспроизводимость и предсказания, но без независимого тканевого референса не доказывают точность абсолютных $\rho_1,\rho_2$.

Для каждого сокращения итогом должна быть область условий, в которой оценена ошибка всех целевых величин. Возможны три решения: допущение допустимо в указанных пределах, недопустимо либо данных для решения недостаточно. Малый остаток и попадание в рабочие физиологические границы необходимы для диагностики, но не заменяют такой проверки. Фиксация неверного значения способна уменьшить разброс оценок и одновременно увеличить их систематическую ошибку.

## 11. Выводы и границы полученного результата

**Наличие КТ позволяет задать геометрию, но точность обратной задачи по-прежнему зависит от её упрощения и от ограничений на тканевые параметры.** В первом опыте ошибка толщины до 1–2 мм вызвала заметное смещение $\rho_2$ при практически точном совпадении импедансов. Следовательно, допущение постоянной толщины нужно оценивать по ошибке сопротивлений в конкретных условиях.

Во втором опыте толщина была точной, однако ограничение на базовую $\rho_1$ исказило дыхательную разность лёгкого. Поэтому устранение геометрической ошибки не делает остальные ограничения автоматически допустимыми. В третьем опыте были точными уже и толщина, и базовые сопротивления, но исключение фонового пульса существенно изменило лёгочную оценку. Постоянство базового уровня и отсутствие пульсового изменения требуют разных доказательств.

Ни один опыт не устанавливает фактическую изменчивость толщины или сопротивлений у человека. Полученные числа характеризуют три заданных нарушения в идеальной плоской модели с точечными электродами. Они также не позволяют ранжировать допущения по опасности для всех измерений: величины нарушений и показатели ошибки в опытах различаются.

Практический результат этого этапа — последовательность проверок, по которой можно обоснованно сокращать модель. Ближайший расчёт должен сравнить фиксированную толщину и восстановление её изменения в принятых пределах, проверить общую базовую $\rho_1$ и оценить передачу базовых ошибок в обе пульсовые составляющие. До проверки их исключения обе пульсовые ткани сохраняются.

Итоговая методика должна задавать место и ориентиры установки, ориентацию и размеры сборок, число и порядок измерений, контакты, дыхательный протокол, внешние входы, обработку сигнала, обратную модель и неопределённость результата. Уменьшение числа сборок и времени расчёта принимается при сохранении установленной точности. Оптимальность относится к проверенному семейству моделей и условиям; универсальная оптимальность из этих опытов не следует. Индивидуальная КТ остаётся текущим входом. Её исключение и установка только по внешним ориентирам относятся к более позднему самостоятельному этапу.
