# 21.09. Независимая разметка камер сердца на исторической контрастной КТ


## Цель и состояние работы

Цель — оценить ошибки объёмов камер и их межфазных изменений, полученных
TotalSegmentator, относительно отдельно выполненной анатомической разметки.
Проверка относится к исторической контрастной КТ. Её результаты нельзя
автоматически переносить на слабоконтрастные серии современных добровольцев.

Подготовлены проекты четырёх фаз для независимой разметки в 3D Slicer.
После разрешения автора выполнен просмотр обезличенных изображений и
сохранены первые агентские черновики видимых частей левого и правого
желудочков одной фазы. Контуры заданы по КТ; маски TotalSegmentator при их
построении не загружались. Интерполяция контуров и порог по интенсивности
применены как явно описанные вспомогательные операции.

Полные проверочные маски камер пока не получены. Сохранённые области
ограничены частью желудочков; их нельзя использовать как полные объёмы,
ударный объём или оценку ошибки TotalSegmentator. Проекты остальных фаз
содержат пустые редактируемые шаблоны. Агентская разметка не проходила
независимое экспертное принятие.


## Выбор материала и анатомическая пригодность

Первоначальный пилот использовал последовательность, обработанную в
[21.08](21.08_Историческая_контрастная_КТ_объёмы_камер.md).
При просмотре её фазы 0% в нижней плоскости изображения обнаружены
продолжающиеся полости желудочков. Это наблюдение указывает на ограниченное
покрытие; оно не позволяет считать рассчитанные объёмы полными.
Первоначальные проекты сохранены с этим ограничением.

Для дальнейшей разметки выбрана другая реконструкция того же архивного
случая. В её фазе 0% нижние плоскости расположены ниже различимой полости
левого желудочка. Выбор сделан по изображению КТ, до просмотра автоматических
контуров этой реконструкции. Подготовлены все 20 фаз; проекты разметки
созданы для 0, 25, 50 и 75%. Эти обозначения отражают шкалу реконструкции,
а не доказанные конечную диастолу и систолу. Полноту каждой камеры во всех
четырёх фазах ещё требуется проверить отдельно.

В первой полной автоматической последовательности маска правого желудочка касается
границы изображения во всех 20 фазах, левого желудочка — в девяти, целого
сердца — в 16. Это вычислительные признаки возможного неполного покрытия или
ошибки сегментации. Они не устанавливают причину и не заменяют просмотра КТ.
Первая задача разметчика — определить, видны ли вся камера, её верхушка,
клапанные границы и переходы в крупные сосуды. Проверку выполняют в трёх
взаимно перпендикулярных плоскостях.

Для каждой камеры в каждой фазе отдельно отмечают полное, частичное или
неопределённое покрытие. Объём видимой части нельзя называть полным объёмом
камеры. Если покрытие недостаточно, соответствующую камеру исключают из
основной оценки полного объёма либо выбирают другую последовательность по
анатомическим критериям. Причину выбора записывают до сравнения чисел.
Недостающую анатомию не достраивают под видом непосредственно наблюдаемой.

Толщина реконструированного среза составляет 5 мм, шаг между соседними
срезами — 2,5 мм. Перекрытие срезов и интерполяция отображения не возвращают
утраченные мелкие детали. Неоднозначность тонких стенок, трабекул и
клапанных границ учитывают отдельно от ошибки программы сегментации.


## Использование контрастирования

По сообщению автора кровь в исторических КТ контрастирована и может быть
близка к кости по яркости. Поэтому интенсивность служит подсказкой при
анатомическом разборе, но не определяет принадлежность вокселя камере.
В подготовленных проектах сохранена отдельная область вокселей с
интенсивностью не ниже 250 HU. Это пробное условие отображения, а не
физиологический диапазон и не установленный порог сегментации крови.
Область включает также кости, сосуды и другие яркие структуры. При открытии
проекта её наложение выключено.

Разметчик может использовать локальный порог, опорные точки и связность как
вспомогательные инструменты. При этом необходимо отделять камеры друг от
друга, от крупных сосудов и миокарда по анатомии. Связность контрастированной
крови через клапанные отверстия сама по себе не задаёт границы камеры.
Пороговую маску без такой проверки нельзя считать ручным референсом.


## Правила анатомической разметки

Перед первой принимаемой разметкой нужно зафиксировать единые правила:

- какие плоскости ограничивают камеры на уровне клапанов;
- где отсекаются аорта, лёгочная артерия, полые и лёгочные вены;
- включаются ли ушки предсердий;
- как учитываются различимые папиллярные мышцы и трабекулы;
- как обозначаются участки, где частичный объём не позволяет провести
  однозначную границу.

Эти решения пока не приняты. Их сохраняют в поле `annotation_policy`
локального файла `workspace.local.json` до фиксации проверочного набора.
Описание решения должно содержать анатомическое правило; одного ответа
«включать» без указания структуры недостаточно. При изменении правила
создают новую версию и отмечают затронутые фазы.

В каждом проекте предусмотрены четыре области: LV — кровь левого желудочка,
RV — правого желудочка, LA — левого предсердия, RA — правого предсердия.
Миокард и крупные сосуды в эти названия не входят. Наружный контур целого
сердца является другим объектом и не выводится из объединения камер.


## Первые сохранённые контуры и результат их проверки

В фазе 0% второй реконструкции контуры желудочков заданы на опорных
аксиальных срезах. Между ними интерполировано поле знакового расстояния
до контура с учётом физических размеров вокселя. Внутри полученной области
рассмотрены три нижних порога: 170, 190 и 210 HU. После порога сохранена
крупнейшая связная трёхмерная область. Все эти операции входят в описание
агентского черновика; они не называются исключительно ручной сегментацией.

Первый вариант левого желудочка был отклонён для оценки полного объёма:
его контур постепенно сужался у основания без установленной клапанной
границы. Такое геометрическое завершение может как отсечь кровь желудочка,
так и захватить соседнюю камеру. Файлы этого варианта сохранены вместе с
причиной отклонения. В текущем варианте явно ограничена только размеченная
часть желудочка. Верхняя граница этой области является границей текущей
работы и не изображает клапан.

Для обоих желудочков сохранены частичные маски, исходные точки контуров,
параметры порога, неопределённые участки, контрольные суммы и редактируемые
сегментации. Просмотрены все аксиальные плоскости размеченной части и два
косых продольных представления. Маски не пересекаются. Эти проверки
устанавливают техническую согласованность и документируют анатомический
просмотр, но не доказывают точность границ.

Отдельно просмотрена тонкосрезовая реконструкция фазы 0%: толщина 0,5 мм,
шаг 0,25 мм. Она содержит более мелкие детали, однако общая система
координат и одинаковая номинальная фаза сами по себе не подтверждают
совпадение анатомии двух реконструкций. Контуры с неё не переносились;
пространственное совмещение не выполнялось. Эта реконструкция не образует
полный тонкосрезовый динамический цикл.

Читательская версия ноутбука показывает наложения, границу частичной
разметки и изменение её объёма при трёх порогах. Изменение числа при смене
порога характеризует зависимость черновика от выбранного способа выделения
крови. Оно не является физиологической вариабельностью, доверительным
интервалом или измеренной ошибкой нейросети.


**Рисунок 1. Границы текущей разметки в двух косых продольных плоскостях.** Зелёный контур показывает размеченную часть левого желудочка, голубой — правого. Прямой верхний край особенно ясно виден справа: разметка заканчивается внутри продолжающейся полости. Это граница выполненной работы, а не найденная клапанная плоскость. Шкалы заданы в миллиметрах вдоль выбранных плоскостей. Порог 190 HU служит одним из исследованных вариантов; его оптимальность не установлена.

**Рисунок 2. Толстосрезовая и тонкосрезовая реконструкции на одинаковых физических плоскостях.** Слева показана реконструкция толщиной 5 мм с шагом 2,5 мм; справа — 0,5 мм с шагом 0,25 мм. Тонкие срезы позволяют различать более мелкую структуру, но изображение не является проверенным пространственным эталоном для левой колонки. Совмещение анатомии отдельно не выполнялось. Эта иллюстрация не обосновывает перенос масок между реконструкциями.

## Порядок работы в Slicer

1. Открыть подготовленный файл запуска нужной фазы. При первом открытии
   области камер пусты. При повторном запуске восстанавливается последняя
   полностью сохранённая версия разметки этой фазы.
2. Проверить полноту изображения камеры в трёх плоскостях. На панели
   «OLD: независимая разметка» установить для каждой камеры состояние
   покрытия: `complete` — полное, `partial` — частичное, `uncertain` —
   неопределённое. `unreviewed` означает, что проверка ещё не выполнена.
3. В модуле Segment Editor выбрать камеру. Использовать Draw, Paint и Erase
   для контуров, закрашивания и исправления. Пороговые и другие подсказки
   применять только внутри анатомически обоснованной области; используемые
   инструменты и существенные параметры записывать в примечания.
4. Проверить все срезы, где присутствует камера, затем границы в остальных
   плоскостях. Трёхмерная поверхность может помогать искать дефекты, но не
   заменяет послойный просмотр. Контуры TotalSegmentator до завершения
   независимой разметки не загружать.
5. Указать разметчика и принятые решения в панели сохранения. Нажать
   «Сохранить НОВУЮ версию разметки». Каждое нажатие создаёт отдельную папку
   с четырьмя NIfTI-масками, редактируемой сегментацией и сведениями о версии.
   Обычное сохранение сцены Slicer не заменяет эту фиксацию.
6. Продолжить остальные фазы с теми же анатомическими правилами. Для
   повторяемости часть фаз повторно размечают независимо от первой версии;
   повторную сессию обозначают в имени разметчика или примечании. Уже
   сохранённые версии не удаляют и не перезаписывают.

Если разметчик видел автоматические контуры вне подготовленного проекта,
это необходимо указать. Отсутствие автоматических масок в проекте уменьшает
риск зависимой разметки, но само по себе не доказывает независимость
решений человека. Агентская разметка, ручная работа человека и экспертное
принятие сохраняются как разные сведения.


## Сохранение и проверка геометрии

Файлы NIfTI экспортируются в исходную сетку КТ с физическими единицами мм.
Slicer может изменять порядок и направление осей массива при загрузке.
Экспорт проверяет, что это только перестановка или разворот индексов,
и обращает их без интерполяции. Произвольные преобразования и изменение
сетки блокируют экспорт до отдельного разбора.

Для каждой версии сохраняются контрольные суммы исходной КТ, масок,
редактируемого проекта и экспортирующего кода; идентификатор фазы; версия
Slicer; имя разметчика; примечания; покрытие камер; число вокселей,
объём и наличие пересечений между камерами. Отсутствующая разметка остаётся
явно отмеченной даже при экспорте пустого файла.

Сохранённая версия имеет статус черновика до проверки. Имя разметчика
не подтверждается программой независимо. Автоматического присвоения
экспертного принятия нет. Изменённый исходный файл КТ, несогласованные
координаты и отсутствие имени разметчика блокируют штатное сохранение версии.

Работоспособность экспорта проверена на синтетическом объёме с неодинаковым
шагом по осям и отражённой системой индексов: подтверждены побитовое совпадение
маски после возврата в исходную сетку, физические координаты, единицы мм,
сохранение исходной КТ и восстановление последней версии. Это проверка кода;
анатомическая точность ею не устанавливается. Использование сегментов и
массива исходного объёма соответствует
[документации 3D Slicer](https://slicer.readthedocs.io/en/latest/developer_guide/script_repository/segmentations.html).


## Сравнение после принятия проверочного набора

В сравнение включают только согласованные по определению камеры, фазы и
геометрии маски. Для каждой фазы рассчитывают ошибку объёма со знаком и её абсолютное значение в мл и процентах, Dice, среднее расстояние между поверхностями
и робастное расстояние Хаусдорфа (HD95) в мм. HD95 определяется по
95-му процентилю расстояний от точек каждой поверхности до другой
поверхности; способ объединения двух направлений фиксируют в коде. Допуск для поверхностного Dice
фиксируют до сравнения; универсальное значение в этом пилоте не принято.
Фазы одного человека не являются независимыми участниками исследования.

Ошибка изменения объёма между двумя фазами определяется как

$$\varepsilon_{\Delta V}=(V_{TS,2}-V_{TS,1})-(V_{ref,2}-V_{ref,1})=e_2-e_1,$$

где $V_{TS}$ и $V_{ref}$ — автоматический и проверочный объёмы в мл,
$e=V_{TS}-V_{ref}$ — ошибка в отдельной фазе. Постоянное смещение может
сократиться при вычитании; фазозависимая ошибка непосредственно искажает
изменение объёма. Поэтому ошибки отдельных объёмов и межфазных разностей
нужно показывать раздельно. Без обоснованных конечных состояний размах
кривой не называется ударным объёмом или фракцией выброса.

Повторяемость разметки и расхождения между разметчиками оцениваются отдельно.
Согласие двух автоматических масок, гладкая кривая или красивый контур
не заменяют проверочный набор. На текущем этапе получены средства разметки и частичные агентские
контуры. Оценка точности TotalSegmentator остаётся следующей задачей.


## Файлы и воспроизводимость

Подготовку выполняет `prepare_old_reference_workspace.py`, создание проектов
и экспорт версий — `slicer_old_reference.py`. Входные абсолютные пути
содержатся только в локальной конфигурации. Пример переносимой конфигурации
находится в `old_reference_workspace.example.json`. Контуры обрабатывает
`render_authored_heart_contours.py`; отчёт из уже сохранённых результатов
создаёт `build_old_reference_report.py`. Его запуск не выполняет
повторную сегментацию и не присваивает принятие маскам.

Реальные проекты, подсказки, сохранённые маски и сведения о разметчиках хранятся
в исключённой из Git локальной папке `results/old_reference_20260915`.
Исходные КТ и маски TotalSegmentator не перезаписываются. В Git сохраняются
код, синтетические тесты, настоящий протокол и научный отчёт. Ни одно сохранение файла
не повышает доказательный статус результата автоматически.


## Следующие действия и условия перехода к сравнению

1. Установить клапанные границы и переходы в выходные тракты на продольных
   и поперечных изображениях. Сохранить точки и правило каждого отсечения;
   участки, где правило неоднозначно, выделить отдельно.
2. Завершить левые и правые желудочки четырёх выбранных фаз. Для каждой
   камеры подтвердить полное покрытие и повторно проверить все её срезы.
   Затем разметить предсердия с явными границами вен и ушек.
3. Проверить повторяемость разметки на повторной независимой сессии.
   Зафиксировать автора принятия и версию проверочного набора. При отсутствии
   принятия сохранять статус черновика и не рассчитывать окончательную
   точность автоматических объёмов.
4. Получить автоматические маски именно второй реконструкции и сопоставить
   их с замороженным проверочным набором. Маски первой реконструкции нельзя
   использовать вместо них без отдельного согласования геометрии и фаз.
5. Рассчитать ошибки объёмов и межфазных изменений раздельно. После пилота
   расширить разметку до всех 20 фаз и других пригодных архивных случаев.
   Разные реконструкции одного человека не считать независимыми участниками.

Документация разработчиков указывает, что модель камер TotalSegmentator
работает на сетке приблизительно 0,73 × 0,72 × 1 мм. Это описание модели;
оно не означает, что при подаче реконструкции толщиной 5 мм появляются
исходно отсутствующие детали. В доступном описании не установлены все
нужные этому сравнению правила клапанных границ и папиллярных мышц.
Их соответствие проверочной разметке остаётся отдельным вопросом.
Источник: [описание модели камер сердца](https://github.com/wasserth/TotalSegmentator/blob/master/resources/heartchambers_highres_details.md).

## Практический итог

Получены редактируемые частичные маски двух желудочков и проверен их экспорт без изменения вокселей и физических координат. Ошибка TotalSegmentator пока не измерена. Следующий содержательный шаг — завершить анатомические границы камер у клапанов и выходных трактов, после чего перейти к другим фазам и независимому принятию разметки.

Локальные данные находятся в папке результатов исследования: проекты четырёх фаз — в `pilot_02`, первые редактируемые черновики — в `body_review/slicer`, исходные точки и варианты масок — в `agent_drafts`. Файл `ОТКРЫТЬ_ЧЕРНОВИКИ.cmd` в `body_review` открывает подготовленную сцену. Эта сцена предназначена для просмотра и исправлений; сохранение в ней не означает принятия референса.
